## Strings and String Manipulation

#### What is a String?

- A String is a Python data type
- A String is a sequence of individual characters, like "c" or "?"
- Characters are arranged in a particular order in a string
- For instance, "tale" and "teal" are different strings, although they contain the same four characters

````python
w1="tale"
w2="teal"
if w1!=w2: print(f"'{w1}' is different than '{w2}'")
````

## Accessing the Individual Characters in a String

- Some programming tasks require you to access the individual characters that are contained within a string.
- You can isolate individual characters within a string by using a "for" loop. The target variable in your loop will assume each character in your string one at a time.
- The following loop prints out each of the characters in "tale" on its own line.

````python
for c in "tale":
  print(c)
````

## Indexing into Strings

- Another way to access a string is to use an index value.
- Index values refer to individual characters
- You use square brackets to index into strings, with the numerical index value inside the brackets
- Index values start at zero for the first character, one for the second, etc.
- You can use negative values to count back from the end

````python
s="Superman"
print(s[0]) # first character "S"
print(s[2]) # third character "p"
print(s[4]) # fifth character "r"
print(s[-2]) # second to last character "a"
print(s[-1]) # last character "n"
print(s[-8]) # first character "S"
# print(s[8]) # this will be out of range
````

## The "len" function

The "len" function will tell us the length of a string.

````python
s="Superman"
print(len(s))
````

If you index into a string with an index more than one minus the length, it will generate an error.

````python
s="Superman"
try:
  print(s[8]) # index should be between 0 and 7, generates an error
except:
  print("indexing error")
````

## Iterating Over a String Using Indexing

As we have seen, you can iterate over a string using the "in" statement and a "for" loop

You can also loop using the index, but you will need to specify a range using the "len" function.

````python
# looping using "in"
word="teal"
for c in word:
  print(c)
````

````python
# looping using "range" and "len" and indexing
# works out the same.
word="teal"
# note if given one parameter, range gives values between
# zero and one minus that parameter
for i in range(len(word)):
  print(i, word[i])
````

## Strings are Immutable

- Strings are an immutable data type. This means that they cannot be changed once they are created.
- This may seem counter-intuitive, since we can (and have) been doing things like the following:

````python
cheese="Cheddar"
print(f"The cheese is {cheese}")
cheese="Brie"
print(f"The cheese is {cheese}")
````

However, what happens in the code above is that Python creates a string "Cheddar" and sets the variable "cheese" to point to it. Then it creates another string "Brie" and changes the variable so it now points to "Brie". The original string "Cheddar" is still in memory but no variable points to it any more.

Since strings are immutable, you can't change the characters in them. The following won't work:

````python
name="Mary"
# try to change it to "Gary"
name[0]="G"
````

## Changing a String

Because strings are immutable, you can't change them by changing the characters in them. You need to construct a new string.

Suppose you want to change every vowel in a word to an underscore. Suppose the word is "hello". You could try to do the following, but it won't work.

````python
word="hello"
word[1]="_"
word[4]="_"
````

Instead, you have to do something like the following:

````python
word="hello"
new_word=""
for c in word:
    if c.lower() in "aeiou": # if is a vowel
      new_word+="_" # put an underscore at the end
    else:
      new_word+=c # put the non-vowel at the end
# we can then set the old variable to the same
# value as the new one
# it now points to the new string
word=new_word
print(word)
````

## The ord Function: Getting the ASCII Value of a Character

- If you recall, Python uses the ASCII coding scheme to store characters.
- The ord() function gives you the ascii value of a character given as its argument
- for instance, ord("A") returns 65

````python
print(ord("A"))
````

## The chr() function: Gets a Character from an ASCII Code

- The chr() function is the reverse of the ORD function
- Given an ASCII code, it returns a character
- For instance, chr(65) returns "A"

````python
print(chr(65))
print(chr(ord("A")))
````

## Extracting Substrings Using "Slicing" Notation

- You can extract substrings of a string by giving the first index and the last index (plus one) desired separated by a colon

````python
full_name="Mirabel Madrigal"
first_name=full_name[:7]
last_name=full_name[8:]
print(first_name)
print(last_name)
s=full_name[8:11:2]
print(s)

l=full_name.split()
print(l)
l="James Earl Jones".split()
print(l)
````

## String Slicing Syntax

- You use brackets to slice a string
- The syntax is as follows:

```text
substring=bigstring[start:end:step]
```

- You must supply at least a starting or ending value
- Substrings start at the start value specified and end at one minus the end value
- Omitting a start value means you start at the beginning
- Omitting a end value means you go all the way to the end
- The syntax is similar to the range() function

````python
# Slicing Examples
s="Superman sings in the shower."
print(s[0:8]) # Superman
print(s[:8]) # Superman
print(s[9:14]) # sings
print(s[:5]) # Super
print(s[9:]) # sings in the shower.
print(s[-7:]) # shower.
print(s[0:len(s):3]) # Seasgit or
````

## Testing for Substrings Using "in" and "not in"

- a in b tests if string a is found inside string b
- a not in b tests for the opposite

````python
names="Pedro Alma Julieta Papa Bruno Augustin Felix Isabela Luisa Mirabel"
print("Luisa" in names) # True
print("Louisa" in names) # False
print("Alma" not in names) # False
print("Ann" not in names) # True
````

## String Testing Methods

- String testing methods allow you to test patterns in a string
- For instance, the "isdigit" method, if given a character, returns True if it is a numeric digit and False otherwise
- If "isdigit" is given a string, it returns true if all of the characters in that string are digits

````python
print("5".isdigit())
print("a".isdigit())
print("1234".isdigit())
print("a123".isdigit())
````

````python
x="3"
y="a3"
print(x.isdigit())
print(y.isdigit())
````

## Some String Testing Methods

- isalnum(): True if all characters are alphanumeric
- isalpha(): True if all characters are alphabetic
- isdigit(): True if all characters are digits
- islower(): True if all alphabetic characters are lower case
- isupper(): True if all alphabetic characters are upper case
- isspace(): True if all characters are "whitespace"

````python
print(ord("\n"))
print("\n".isspace())
print(" ".isspace())
print("\t".isspace())
print("\n\n\n")
s="Hello\nThere\nWorld!"
print(s)
````

## String Modification Methods

- lower(): returns a lowercase version of a string
- upper(): returns an uppercase version of a string
- rstrip(): removes whitespace at the end of a string
- lstrip(): removes whitespace at the beginning of a string
- strip(): removes whitespace at both the beginning and end of a string
- capitalize(): capitalizes a string (first character upper case)
- title(): capitalizes the first character of each word in a string
- swapcase(): turns lower case to upper and vice versa in a string

````python
print("HELLO".lower()) # hello
print("hello".upper()) # HELLO
print("hello   ".rstrip())
print("  hello".lstrip())
print("  hello  ".strip())
print("hello".capitalize()) # Hello
print("hello there".title()) # "Hello There"
print("Hello There".title().swapcase()) #  "hELLO tHERE"
````

## Finding Substrings

- You can find the location of a string inside another string using the find() method
- It will return the index of the first occurence of the substring within a string.
- If it isn't found, it will return -1

````python
sentence="Like finding a needle in a haystack!"
location=sentence.find("needle")
print(location)
l2=sentence.find("sword")
print(l2)
````

## Replacing Substrings

- The replace() method will replace all occurences of a substring in a string with a new string
- An example is the following:

````python
text="""Voldemort had one goal in life:
to kill Harry Potter.
Voldemort hated Harry."""
print(text)
newtext=text.replace("Voldemort","He who should not be named")
print(newtext)
````
