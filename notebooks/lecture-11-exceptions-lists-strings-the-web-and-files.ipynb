## Exceptions

An exception is any error that that causes a program to stop while it is running.

For instance, converting a string not containing digits to an integer will generate an error, because it can't be done (it doesn't make sense if the string isn't all digits.)

````python
x="Hello"
#x="123"
y=int(x)
print(y)
````

- When an exception occurs Python will generate a “traceback," as it did above
- Tracebacks give information about the exception that occurred, including the line number that caused the issue and the name of the exception
- Programmers generally say that an exception has been “raised” by the program

## Preventing Exceptions

Many times, you can avoid exceptions by changing our code. For example, the following code will raise an exception if x is zero.

````python
x=int(input("Please enter a number:"))
print(100/x)
````

However, we can avoid the issue by wrapping the potentially dangerous code inside a conditional statement:

````python
x=int(input("Please enter a number:"))
if x!=0: print(100/x)
````

## Catching Exceptions

- Python has an exception handling statement that may be used to "catch" exceptions and prevent them from crashing your program.
- It uses the following syntax:

```text
try:  # questionable code goes hereexcept:  # this block will run if the code in the "try"  # block above raised an exceptionelse: # optional  # this block will run if the code in the "try"  # block above was successful    
```

````python
# catch the division by zero error
x=int(input("Please enter a number:"))
try:
  print(100/x)
except:
  print("Can't divide by zero.")
````

## Splitting a String

- Sometimes you are given long strings of information that need to be "parsed" in a particular way
- For example, consider the following string:

```text
x = "Adams,Mayor,NYC"
```

- This string contains three pieces of data: a name, a title and an employer. The data has been "packaged" into a single string by using a comma to separate the different data items.
- We can use a technique called "splitting" to "unpack" a string and extract its individual values. The trick is to use a separator value, a "delimiter".
- The split method splits a string into a list of strings using a delimiter.

````python
x = "Adams,Mayor,NYC"
splitstring=x.split(",")
print(splitstring)
````

## Obtaining Data from the Web

- You generally won't be "hard coding" data into your programs. You will usually be processing information that comes to you in the form of user input or via an external data source.
- One way to easily to get information into Python from the "outside world" is to initiate a connection to the Web and obtain data via a URL.

Method:

- Initiate a web connection to an external data source
- Obtain the data that exists at that location as a String
- "Clean up" the string into a meaningful form that your program can understand. This usually involves making sure that the character set used in the external file is one that your computer can handle
- Parse your string into a list
- Process this list and generate output

````python
# an example: getting a web page from CNN using the urllib module
import urllib.request
# source of data:
url = 'http://www.cnn.com'

response = urllib.request.urlopen(url)

data = response.read().decode('utf-8')

print(data[:2000])
````

## Be Careful!

- The web is a volatile place: servers can go up and down at a moment's notice, and your connection to the network may not always be available.
- You need to be prepared to deal with connection errors in your programs.
- You can use the try/except/else statement to help catch errors before they crash your program

## Programs and Memory

- All of the programs that you have been writing so far have reset themselves each time they run
- This is because the data associated with the program (i.e. your variables) are stored in your computer’s memory (RAM)
- RAM is a volatile place – it serves as a computer’s short term memory. RAM that is being used by a program is cleared when the program stops running.
- If programs are to retain data between executions then we must find a more permanent way to store and save information for future use.

## Long Term Storage

- A computer almost always has at least one type of long-term storage device at its disposal (usually some kind of hard drive, SSD, or cloud storage)
- We can use the long term storage capabilities of a computer to store data in the form of a file.
- Once we save a file it will remain on the long term storage device after the program is finished running, and can be accessed and retrieved later on.
- Many applications use long term storage: e.g. word processors, spreadsheets, Adobe Acrobat for PDFs, photo display programs, etc.

## File Types

There are two main file types:

- Text files (human readable) (this notebook is a text file)
- Binary files (readable only by the machine, like photo files)

## Text Files

- Text files are files that contain data that is encoded as text (i.e. ASCII or Unicode characters)
- Data stored in a text file is visible and can be read by a program that is designed to view / edit textual data (i.e. a word processing program)
- Examples:
  - Notepad documents
  - Cookies
  - HTML files
  - XML files

We will be working with text files exclusively in this class

## Binary Files

- Binary files contain data that is not encoded in a text format
- Binary files are intended to be read by other programs and not by humans directly
- Binary files appear as “gibberish” when viewed via a word processing program
- Examples
  - Image files (PNG / GIF / JPG files)
  - Movie Files (MOV / MP4 / AVI files)
  - Application files (EXE / APP files)

## Working with Files

- Step 1: Open up a connection between your program and the operating system. Define the name and location of the file you wish to work with as well as what you want to do to that file (read or write)
- Step 2: Process the data
  - Writing: send information in the form of variables from your program into the file to be written
  - Reading: extract information into variables in your program
- Step 3: Close the file, telling the OS to release the file for use by other programs

## File Access Methods

- Sequential Access: You must read a file starting at the beginning and continue until you reach the data you are looking for (similar to how a cassette tape works) – this is the method we will be focusing on in this class
- Random Access: You can skip from point to point in a file, allowing you to jump over data until you find what you are looking for (similar to how a CD works)

## Filenames

- Filenames are nothing more than labels that allow us to uniquely identify groupings of data that exist on a long term storage device
- Every file must have a filename
- On most operating systems a filename comes with a “file extension” which allows you to quickly tell what kind of data is stored inside that file. Extensions come after the last period in a file (i.e. “mydocument.doc” – “doc” is the file extension)
- File extensions tell the operating system what kind of data is stored in the file and allows the OS to select an appropriate program to open a particular file

## File Objects

- A file object is a special variable that you must create in a programming language that serves as a connection between your program and the operating system (OS)
- File objects tell the OS the name of a file that you wish to work with along with what you want to do to that file (write, read, append, etc)

## Opening a File in Python

```text
file_object = open(filename, mode)# File Modes# 'r' = read# 'w' = write - if the file does not exist, create it#               if the file does not exist, overwrite it# 'a' = append - if the file does not exist, create it#               if the file does exist, append data to the end of it
```

- Filenames are expressed as strings
- If you type in a simple filename (i.e. “myfile.txt”) you are telling Python to open up a file with that name in the current directory
- You can ask Python to open up a file that exists outside of the current directory by typing in an absolute path. For example:
  - Windows: ‘C:\temp\testfile.txt’
  - Mac or Linux/Unix: ‘/var/local/testfile.txt’
- When providing an absolute path you should preface the string with the ‘r’ directive – this tells Python to treat the text in the string as “raw” text and not try and add in any escape characters. Example:
  - file\_object = open(r’C:\temp\testfile.txt’, ‘w’)

## Delimiters and Files

- You want to try and avoid writing files that concatenate all your data into one long line of unintelligible text
- This is bad practice since it can be very difficult – or impossible – to extract out your data later on
- One way to separate data in a text file is by separating your data into multiple lines using the “\n” escape character. This allows you to store a single value on each line of your text file, which will make things a lot easier for you later, when you need to read your file and perform an operation on the data contained within

The file worldseries.txt can be found [here](https://www.google.com/url?q=https%3A%2F%2Fdocs.google.com%2Fdocument%2Fd%2F1jaZrFGI-xXAD_NAiz9x6iha3supuuIgI60k0t12IPH4%2Fedit%3Fusp%3Dsharing).

It has one line for each year for each winner of the World Series in baseball up to a few years ago. Let's determine what team won the World Series the most times.

It uses my Google Drive. If you want to run it, you will need to copy the file into your own Google Drive and [mount your Google Drive into Colab](https://colab.research.google.com/notebooks/io.ipynb#scrollTo=u22w3BFiOveA). Or you can download the file into a directory on your computer from this repo and/or run the program using IDLE or another IDE.

````python
# open the file
# note this references my Google Drive (and a subdirectory of a directory of it)
# to run this, you will have to copy the file to your Google Drive as
# you won't be able to access mine
# you will also have to mount your Google Drive into Colab
f=open("worldseries.txt","r")
# read the entire file
allLines=f.read()
f.close()
# split into lines
teamList=allLines.split("\n")
# following is list of unique teams
tl2=[]
# following is the win count as a list
counts=[]
for team in teamList:
  # if the team is not in the list
  if team not in tl2:
    tl2.append(team) # add it
    counts.append(1) # set the win count to one
  else:
    i=tl2.index(team) # otherwise find its position in the list
    counts[i]+=1 # and increment its count
m=max(counts) # find maximum wins
i=counts.index(m) # find index of maximum wins
print(f"The {tl2[i]} won {m} times.")# print corresponding team and number of wins
print(tl2) # print the list of teams
print(counts) # print all of the counts
````
