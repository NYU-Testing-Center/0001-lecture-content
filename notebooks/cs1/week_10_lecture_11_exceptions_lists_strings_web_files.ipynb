<p>An exception is any error that that causes a program to stop while it is running.</p>
<p>For instance, converting a string not containing digits to an integer will generate an error, because it can't be done (it doesn't make sense if the string isn't all digits.)</p>
<ul>
  <li>When an exception occurs Python will generate a “traceback," as it did above</li>
  <li>Tracebacks give information about the exception that occurred, including the line number that caused the issue and the name of the exception</li>
  <li>Programmers generally say that an exception has been “raised” by the program</li>
</ul>
</article>

  <h2 id="ed-slide-662525-heading">Obtaining Data from the Web</h2>
<ul>
  <li>You generally won't be "hard coding" data into your programs. You will usually be processing information that comes to you in the form of user input or via an external data source.</li>
  <li>One way to easily to get information into Python from the "outside world" is to initiate a connection to the Web and obtain data via a URL.</li>
</ul>
<p>Method:</p>
<ul>
  <li>Initiate a web connection to an external data source</li>
  <li>Obtain the data that exists at that location as a String</li>
  <li>"Clean up" the string into a meaningful form that your program can understand. This usually involves making sure that the character set used in the external file is one that your computer can handle</li>
  <li>Parse your string into a list</li>
  <li>Process this list and generate output</li>
</ul>
</article>

```python
# an example: getting a web page from CNN using the urllib module
import urllib.request
# source of data:
url = 'http://www.cnn.com'

response = urllib.request.urlopen(url)

data = response.read().decode('utf-8')

print(data[:2000])
```

  <h2 id="ed-slide-662530-heading">Text Files</h2>
<ul>
  <li>Text files are files that contain data that is encoded as text (i.e. ASCII or Unicode characters)</li>
  <li>Data stored in a text file is visible and can be read by a program that is designed to view / edit textual data (i.e. a word processing program)</li>
  <li>Examples:

Notepad documents

Cookies

HTML files

XML files</li>
</ul>
<p>We will be working with text files exclusively in this class</p>

  <h2 id="ed-slide-662531-heading">Binary Files</h2>
<ul>
  <li>Binary files contain data that is not encoded in a text format</li>
  <li>Binary files are intended to be read by other programs and not by humans directly</li>
  <li>Binary files appear as “gibberish” when viewed via a word processing program</li>
  <li>Examples

Image files (PNG / GIF / JPG files)

Movie Files (MOV / MP4 / AVI files)

Application files (EXE / APP files)</li>
</ul>
</article>

```python
# open the file
# note this references my Google Drive (and a subdirectory of a directory of it)
# to run this, you will have to copy the file to your Google Drive as
# you won't be able to access mine
# you will also have to mount your Google Drive into Colab
f=open("/content/drive/MyDrive/0002/Data/worldseries.txt","r")
# read the entire file
allLines=f.read()
f.close()
# split into lines
teamList=allLines.split("\n")
# following is list of unique teams
tl2=[]
# following is the win count as a list
counts=[]
for team in teamList:
  # if the team is not in the list
  if team not in tl2:
    tl2.append(team) # add it
    counts.append(1) # set the win count to one
  else:
    i=tl2.index(team) # otherwise find its position in the list
    counts[i]+=1 # and increment its count
m=max(counts) # find maximum wins
i=counts.index(m) # find index of maximum wins
print(f"The {tl2[i]} won {m} times.")# print corresponding team and number of wins
print(tl2) # print the list of teams
print(counts) # print all of the counts
```
