  <h2 id="ed-slide-662621-heading">Object-Oriented Programming with Programmer-Defined Objects and Classes in Python</h2>
<p>Everything in Python is an object, and every variable in Python points to an object which is an instance of a class.</p>
<p>You can think of the class as a set and objects in that class as members of that set. For instance, a and b are both objects that are instances of the class "string."</p>
<p>c and d are both instances of the class "list."</p>
</article>

```python
# All objects have methods that are specific to that type
s="the rain in Spain"
print(s.lower())
x=1.5
try:
  print(x.lower())
except:
  print("can't apply the 'lower' method to a float")
#print(x.lower())
```

```python
a="hello"
print(type(a)) # str
b="world"
print(type(b)) # str
y=2.6
print(type(y)) # float
z=5
print(type(z)) # int
def addOne(x):
  return x+1
z=addOne(z)
print(z)
print(type(addOne)) # function
```

  <h2 id="ed-slide-662623-heading">Python Programmer-Defined Objects and Classes</h2>
<h3>Terminology</h3>
<ul>
  <li>A class defines a type of object</li>
  <li>An object is an instance of a class</li>
  <li>When defining a class, we define attributes and methods of that class</li>
  <li>An attribute is a variable associated with a class and its objects</li>
  <li>A method is a function associated with a class and its objects</li>
</ul>
<h3>We can define our own classes and then create objects using them.</h3>
<h3>Syntax of Programmer-Defined Classes</h3>
<ul>
  <li>"Class" keyword defines a class giving the class name</li>
  <li>Methods are defined as functions indented under the class</li>
  <li>"Self" refers to the object itself</li>
  <li>All methods refer to self and optionally some other arguments</li>
  <li>Special "dunder" or magic methods begin and end with double underscore</li>
</ul>
</article>

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-long" aria-labelledby="ed-slide-662624-heading">
  <h2 id="ed-slide-662624-heading">Class methods</h2>
<ul>
  <li>__init__ method is used to create an object</li>
  <li>__str__ method is used to tell the print function how to print an object</li>
  <li>__eq__ tells the == operator when two objects are equal</li>
  <li>Methods are called with dot notation</li>
  <li>Objects are created with the class name and the arguments to __init__</li>
</ul>
<p>A class for dogs with four methods:</p>
</article>

```python
# define a class
class Dog:
  def __init__(self,name,weight,age,got_rabies_shot):
    self.name=name
    self.weight=weight
    self.age=age
    self.got_rabies_shot=got_rabies_shot
  def __str__(self):
    s=self.name+" is a dog that weighs "
    s+=str(self.weight)+" pounds and is "
    s+=str(self.age)+" years old."
    if self.got_rabies_shot:
      s+="\nVaccinated for Rabies."
    else:
      s+="\nNot vaccinated for Rabies."
    return s
  def have_birthday(self):
    self.age+=1
  def compare_age(self, other):
    if self.age>other.age:
      print(self.name,"is older.")
    elif self.age<other.age:
      print(other.name,"is older.")
    else:
      print(self.name,'and',other.name,'are the same age.')

# create an instance (calls __init__)
dog1=Dog("Fido",25,1,False)
# Dog.__init__(dog1,"Fido",25,1,False)
# print out the attributes
print(dog1.name)
print(dog1.weight)
print(dog1.age)
print(dog1.got_rabies_shot)
print(dog1)
dog2=Dog("Spot",33,5,False)
print(dog2)
dog2.compare_age(dog1)
Dog.compare_age(dog2,dog1)
dog1.compare_age(dog2)
Dog.compare_age(dog1,dog2)


# change dog1's name
dog1.name="Rover"
# print the object (calls the __str__ method)
print(dog1)
# if he has a birthday he will be one year older
dog1.have_birthday()
print(dog1)
```

  <h2 id="ed-slide-662626-heading">A Point Class</h2>
<p>The following is a class that defines points in the X-Y plane.</p>
<p>It has four methods:</p>
<ul>
  <li>A "dunder" __init__ method which tells how a point is initialized (with x and y values)</li>
  <li>A "dunder" __str__ method which tells how a point is printed</li>
  <li>A "dist_to_origin" method which tells how far a point is from the origin.</li>
  <li>A "manhattan_distance" method which tells the "Manhattan Distance" between two points, which is the distance between them using only horizontal and vertical motion in the x-y grid.</li>
</ul>
<p>It also has a "doc string" which is the string right below the class definition.</p>
</article>

```python
class point:
  '''Defines a point in terms of its x and y coordinates'''
  def __init__(self, x, y):
    self.x=x
    self.y=y
  def __str__(self):
    return f"({self.x}, {self.y})"
  def dist_to_origin(self):
    return (self.x**2+self.y**2)**(0.5)
  def manhattan_distance(self,other):
    return abs(self.x-other.x)+abs(self.y-other.y)

p1=point(3,2)
print(p1)
p2=point(1,1)
print(p2)
print(p2.dist_to_origin())
print(p1.manhattan_distance(p2))
print(p2.manhattan_distance(p1))
print(point.__doc__)
```

  <h2 id="ed-slide-662629-heading">A Square Class</h2>
<p>Following is a a class that defines a square in terms of its lower left corner x, y coordinates and the length of its side.</p>
<p>It has the __init__ method and one additional method, "upper_right", which finds the point at the upper right of the square.</p>
<p>It also has a doc string.</p>
</article>

```python
class point:
  '''Defines a point in terms of its x and y coordinates'''
  def __init__(self, x, y):
    self.x=x
    self.y=y
  def __str__(self):
    return f"({self.x}, {self.y})"
  def dist_to_origin(self):
    return (self.x**2+self.y**2)**(0.5)
  def manhattan_distance(self,other):
    return abs(self.x-other.x)+abs(self.y-other.y)

class square:
  '''Defines a square in terms of the x and y
  coordinates of its lower left corner and the
  length of a side'''
  def __init__(self,llx,lly,side):
    self.ll=point(llx,lly)
    self.side=side
  def upper_right(self):
    return point(self.ll.x+self.side,self.ll.y+self.side)


sq1=square(0,0,2)
print(sq1.ll)
print(sq1.upper_right())
print(square.__doc__)
```

  <h2 id="ed-slide-662637-heading">A Random Playing Card Class</h2>
<p>Following is a class that deals a random card from a deck (with replacement, which means you can get the same card twice.)</p>
<p>It has a __str__ method for printing, and methods which tell whether two cards have the same suit or are the same card. The latter is the __eq__ method which defines the "==" operator for the class.</p>
</article>

```python
class random_card:

  def __init__(self):
    suits=("Hearts","Diamonds","Spades","Clubs")
    values=("Ace","2","3","4","5","6","7","8","9","10","Jack","Queen","King")
    import random
    self.suit=random.choice(suits)
    self.value=random.choice(values)
  def __str__(self):
    return self.value+" of "+self.suit
  def same_suit(self,other):
    return self.suit==other.suit
  def __eq__(self,other):
    return (self.suit==other.suit) and (self.value==other.value)

card1=random_card()
print(card1)
card2=random_card()
print(card2)
print(card2.same_suit(card1))
print(card1.same_suit(card2))

# Let's see how long it takes to get two cards that are identical.
def findMatchingPair():
  i=0
  while True:
    i+=1
    card1=random_card()
    card2=random_card()
    if card1==card2: return i
print(f"It took {findMatchingPair()} pairs of cards to get the same card twice.")

total=0
for i in range(1000):
  total+=findMatchingPair()
average=total/1000
print(average)
```
