  <h2 id="ed-645915-what-is-a-computer-heading">What is a Computer?</h2>

  <h2 id="ed-645917-alan-turing-and-the-turing-machine-heading">Alan Turing and the Turing Machine</h2>
<p>British mathematician Alan Turing (1912-1954) developed many of the ideas that led to the development of the modern computer</p>
<p>One of his main conceptions was the Turing Machine, which consisted of a “head” which reads and writes symbols from/to the tape, which moves backwards and forwards so that different parts of the tape. Turing proved that the Turing Machine could execute a very wide range of programs.</p>
<p>Turing influenced John Von Neumann, one of the leading early computer designers.</p>
<p>Turing had a tragic life because he was treated horribly by the British government because he was gay, despite his invaluable services in code-cracking during WWII</p>
<p>Below is a model of a Turing Machine:</p>
<figure class="lecture-ed-figure" style="margin:1rem auto 0;text-align:center;"><img src="images/lecture_01/645917_turing_machine_model.png" alt="A physical model of a Turing Machine with two tape reels." style="display:block;height:auto;margin:0 auto;max-height:42vh;max-width:100%;object-fit:contain;width:auto;"></figure>

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-reading-slide  " aria-labelledby="ed-646274-processor-cpu-heading">
  <h2 id="ed-646274-processor-cpu-heading">The Processor (Central Processing Unit or CPU)</h2>
<ul>
  <li>Every computer has at least one processor and most have several processors</li>
  <li>The processor acts as the computer’s “brain” and coordinates all its operations.</li>
  <li>A processor can perform four distinct operations (typically within one or more nanoseconds):</li>
</ul>
<ul>
  <li>Receive a new instruction (fetch)</li>
  <li>Make sense of this instruction (decode)</li>
  <li>Perform the action defined by this instruction (execute)</li>
  <li>Store the result of the action (store)</li>
</ul>
<p>A processor is made up of miniaturized transistors and other components and circuits on a semiconductor “chip.” The following shows a microprocessor chip next to a penny.</p>
<figure class="lecture-ed-figure" style="margin:1rem auto 0;text-align:center;"><img src="images/lecture_01/646274_microprocessor_and_penny.png" alt="An Intel microprocessor next to a United States penny for scale." style="display:block;height:auto;margin:0 auto;max-height:42vh;max-width:100%;object-fit:contain;width:auto;"></figure>
</article>

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-text-slide  " aria-labelledby="ed-646275-moores-law-heading">
  <h2 id="ed-646275-moores-law-heading">Moore's Law</h2>
<ul>
  <li>Advances in microchip design have facilitated the development of computer technology thanks to Moore’s Law. Gordon Moore, one of the founders of Intel, first formulated the law.</li>
  <li>Moore’s Law is the empirical observation that the complexity of an integrated circuit (microchip) doubles about every two years.</li>
  <li>It’s unclear how much longer Moore’s Law will continue to describe the state of affairs, but it has done so in the past.</li>
  <li>Moore’s law is why computers have gotten faster and faster over time.</li>
</ul>
</article>

  <h2 id="ed-646277-computer-memory-ram-heading">Computer Memory (RAM)</h2>
<ul>
  <li>Computers, like us, need short-term memory in order to function properly</li>
  <li>We call this “Random Access Memory” or RAM and it is very fast!</li>
  <li>We call it “random” since the computer can selectively read and write to any place in its RAM in a fixed (short) amount of time</li>
  <li>When a computer turns off, it loses what is stored in RAM (we call this volatility)</li>
  <li>In general, the more RAM a system has, the better it performs. Often the amount of RAM is more important to performance than CPU speed, although that depends on the application.</li>
</ul>
<p>Here's a picture of some RAM "sticks" with memory chips on them.</p>

  <h2 id="ed-646278-long-term-computer-storage-heading">Long Term Computer Storage</h2>
<ul>
  <li>Temporary storage is accomplished in memory (RAM)</li>
  <li>We use RAM for short term memory because it is very fast, but it is also volatile and loses it contents when the computer is shut off</li>
  <li>Long term storage is, by definition, non-volatile- it has the ability to persist even if the power is off</li>
  <li>It also typically holds far more data than the RAM can</li>
  <li>Long term storage devices include hard drives, solid-state drives, and USB “key” drive</li>
  <li>Sometimes this storage is available in the cloud</li>
</ul>
<p>Here is a picture of a hard drive:</p>

  <h2 id="ed-646287-computers-only-understand-binary-codes-heading">Computers Only Understand Binary Codes</h2>
<ul>
  <li>The underlying language, the machine language, of a computer is numerical and is expressed in binary (zeros and ones).</li>
  <li>Each code represents an instruction to the CPU (e.g. add two numbers)</li>
  <li>High-level languages, like Python and Java, are translated into machine language by a program called an interpreter or a compiler</li>
  <li>Python is usually an interpreted language but compilers exist</li>
  <li>Compilers produce faster code than interpreters</li>
</ul>

<p>A First Program</p>
<p>The program on the right converts temperature from Fahrenheit to Celsius.</p>
<p>The program uses the input function to get a value of Fahrenheit degrees from the user. It then converts that input to a floating point number using the float function. It stores the result in the degreesF variable. It then converts degreesF to degreesC using a formula, and prints out the result.</p>

```python
degreesF=float(input("How many degrees Fahrenheit?"))
degreesC=(degreesF-32)*5/9
print("That is", degreesC,"degrees Celsius")
```
