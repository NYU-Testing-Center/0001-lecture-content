  <h2 id="ed-slide-662259-heading">Changing a String</h2>
<p>Because strings are immutable, you can't change them by changing the characters in them. You need to construct a new string.</p>
<p>Suppose you want to change every vowel in a word to an underscore. Suppose the word is "hello". You could try to do the following, but it won't work.</p>
<p>Instead, you have to do something like the following:</p>
</article>

```python
word="hello"
word[1]="_"
word[4]="_"
```

```python
word="hello"
new_word=""
for c in word:
    if c.lower() in "aeiou": # if is a vowel
      new_word+="_" # put an underscore at the end
    else:
      new_word+=c # put the non-vowel at the end
# we can then set the old variable to the same
# value as the new one
# it now points to the new string
word=new_word
print(word)
```

  <h2 id="ed-slide-662274-heading">String Modification Methods</h2>
<ul>
  <li>lower(): returns a lowercase version of a string</li>
  <li>upper(): returns an uppercase version of a string</li>
  <li>rstrip(): removes whitespace at the end of a string</li>
  <li>lstrip(): removes whitespace at the beginning of a string</li>
  <li>strip(): removes whitespace at both the beginning and end of a string</li>
  <li>capitalize(): capitalizes a string (first character upper case)</li>
  <li>title(): capitalizes the first character of each word in a string</li>
  <li>swapcase(): turns lower case to upper and vice versa in a string</li>
</ul>
</article>

```python
print("HELLO".lower()) # hello
print("hello".upper()) # HELLO
print("hello   ".rstrip())
print("  hello".lstrip())
print("  hello  ".strip())
print("hello".capitalize()) # Hello
print("hello there".title()) # "Hello There"
print("Hello There".title().swapcase()) #  "hELLO tHERE"
```
