  <h2 id="ed-slide-655871-heading">Decision Structures (If--Elif-Else) and Boolean Logic</h2>
<p>So far we have been writing programs that execute one step after another. But it is often necessary to have programs that take different actions based on the data input to them. This is done with a decision structure: an "If" statement.</p>
</article>

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-long" aria-labelledby="ed-slide-655874-heading">
  <h2 id="ed-slide-655874-heading">An Example: Overtime Pay</h2>
<p>For instance, in the U.S., many workers are legally entitled to extra pay if they work more than 40 hours in a week. This is often paid as "time and a half" which means that they get 1.5 times their hourly wage for every hour over 40.</p>
<p>For instance, if your hourly wage is $20, your overtime wage is $30. If you work 60 hours in a week, you earn 40 * $20 == $800 plus 20 * $30 == $600, for a total of $1400.</p>
</article>

```python
hours=float(input("How many hours did you work last week?"))
wage=float(input("What was your hourly wage?"))
if (hours>40.0):
  # calculate non-overtime and overtime pay
  non_overtime=40*wage
  overtime=(hours-40)*(wage*1.5)
if (hours<=40.0):
  # otherwise no overtime
  non_overtime=hours*wage
  overtime=0
total_pay=non_overtime+overtime
print(f"Your pay should be ${total_pay:.2f} of which ${overtime:.2f} was overtime pay.")
```

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-image-slide  " aria-labelledby="ed-slide-655875-heading">
  <h2 id="ed-slide-655875-heading">"If" statements</h2>
<p>"If" statements are an example of a selection statement. In the overtime example, the program selects and executes the first group of statements after the "if" when the expression following the first "if" is true and the second group when the expression following the second "if" is true. Since only one can be true, only one of these groups of statements are executed.</p>
<p>The expression after the "if" is called a Boolean expression because it takes on a value of True or False. Here the expression following the first "if" is "(hours&gt;40)". This will be True if the variable "hours" (input by the user) is greater than 40 and False otherwise.</p>
<p>Boolean expressions are named after the nineteenth-century English mathematican George Boole who developed <a class="amber-el amber-link amber-el ed-focus-outline" href="https://www.google.com/url?q=https%3A%2F%2Fen.wikipedia.org%2Fwiki%2FBoolean_algebra" target="_blank" rel="noopener noreferrer">Boolean algebra</a>, which defined the True and False values and the AND, OR and NOT operations that underpin modern computing and the logic used in all computer languages, including Python. In the computer, True is represented by a value of 1 and False with a value of 0.</p>

</article>

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-short" aria-labelledby="ed-slide-655876-heading">
  <h2 id="ed-slide-655876-heading">Boolean Expressions</h2>
<ul>
  <li>Used as the condition in the "If" statement and other statements that do logical tests that we will see later on (e.g. "while")</li>
  <li>Usually formed using "relational operators" like less than ("&lt;") or greater than which allow you to compare two values and get a result.</li>
</ul>
</article>

```python
hours=float(input("How many hours did you work last week?"))
# (hours>40) is a Boolean expression
# overtimeFlag is a Boolean variable which takes on the value of True or False
overtimeFlag=(hours>40)
print("Did you work overtime?",overtimeFlag)

if overtimeFlag:
  print("You worked overtime.")
else:
  print("You didn't work overtime.")
```

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-short" aria-labelledby="ed-slide-655877-heading">
  <h2 id="ed-slide-655877-heading">Relational Operators</h2>
<p>Produce a value of True or False</p>
<ul>
  <li>a&gt;b (a is greater than b)</li>
  <li>a&lt;b (a is less than b)</li>
  <li>a==b (a is equal to b) (different from the assignment operator "=")</li>
  <li>a!=b (a is not equal to b)</li>
  <li>a&lt;=b (a is less than or equal to b)</li>
  <li>a&gt;=b (a is greater than or equal to b)</li>
</ul>
</article>

```python
ageSimon=int(input("Enter Simon's age:"))
ageSasha=int(input("Enter Sasha's age:"))
if (ageSimon>ageSasha):
  print("Simon is older than Sasha")
else:
  print("Either Sasha is older than Simon, or they are the same age.")
```

  <h2 id="ed-slide-655896-heading">If-Elif-Else</h2>
<p>Elif is a way to express additional conditions on the same level as the first "if". Elif is short for "else if." An elif is only run if all the previous conditions have failed to be True. The final "else" is only run if all of the previous conditions have failed. You can have as many elifs as you like.</p>
<p>Elifs aren't necessary, because everything could be written with nested ifs as shown previously, but they can simplify code by eliminating the need for nesting.</p>
</article>

```python
weight=float(input("Enter the weight of your dog in pounds:"))
if weight<=5:
  dogType="teacup"
elif weight<=12:
  dogType="toy"
elif weight<=24:
  dogType="small"
elif weight<=57:
  dogType="medium"
else:
  dogType="large"
print(f"You have a {dogType} dog.")
if (weight>=1000):
  print("Is it Clifford?")
```

  <h2 id="ed-slide-655903-heading">Generating Random Numbers and then Guessing them</h2>
<p>To generate random integers, you first import the random module and then you use the randint method. The randint method takes the two limits of the range of random integers you want (inclusively.)</p>
<p>This program picks a random number and then sees if the user can guess it.</p>
</article>

```python
import random
# generate a random integer between 1 and 3
secret=random.randint(1,3)
guess=int(input("What is your guess between 1 and 3?"))
if (guess==secret):
  print("You guessed it!")
else:
  print("Sorry, you didn't guess it.")
```
