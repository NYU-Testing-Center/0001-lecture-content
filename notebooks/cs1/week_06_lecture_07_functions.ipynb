  <h2 id="ed-slide-659771-heading">Python Functions</h2>
<ul>
  <li>A function is a group of statements that exist within a program for the purpose of performing a specific task</li>
  <li>We have already been using a number of Python's built in functions, including:

range(): generates a range of integers

print(): outputs data to the console

len(): measures the length of an object, like a list or a string

random.randint(): generates a random integer within a range</li>
</ul>
</article>

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-short" aria-labelledby="ed-slide-659772-heading">
  <h2 id="ed-slide-659772-heading">Defining Functions</h2>
<ul>
  <li>Functions, like variables, must be named and created before you can use them</li>
  <li>The same naming rules apply to both variables and functions:

You can't use any of the keywords Python reserves to itself, like "if", "while", or "for"

No spaces in a name

First character must be in the alphabet, or be the underscore character

After the first character, you can use alphabetic or numeric digits or the underscore

Python is case-sensitive, which means, for instance, that "A" and "a" are not the same</li>
</ul>
<p>The following is an example function definition, "myfunction". Note we use the def keyword to define the function; to call it we refer to it by its name.</p>
</article>

```python
# define a function
def myfunction():
  print("Hello, I'm in the function!")

# call it twice with an intervening print
myfunction()
print("Hi, I'm in the main program.")
myfunction()
```

  <h2 id="ed-slide-659773-heading">Some Notes on Functions</h2>
<ul>
  <li>To run a function, you "call" it. This transfers control up to the function and the statements indented under the function are run.</li>
  <li>Once a function has completed, Python continues with the line after the call.</li>
  <li>Functions must be defined before they are called. We generally place all of our functions at the beginning of our programs.</li>

<p>The following is an example that calls two functions.</p>

  <h2 id="ed-slide-659774-heading">Local Variables</h2>
<p>Local variables are variables that exist only inside the function (inside its "scope," which is the code block associated with it) and do not exist outside of it. This is best seen with an example.</p>
</article>

```python
def bugs():
  numbugs=int(input("How many bugs?"))
  print(f"You entered {numbugs} bugs.")

bugs()
# if you try to print the variable numbugs here, it won't work
# since it is defined only in the function and is local to it
#print(numbugs)
```

  <h2 id="ed-slide-659775-heading">Something Weird</h2>

  <h2 id="ed-slide-659777-heading">Don't Re-use the Same Variable Name</h2>
<p>Don't confuse yourself and others by giving different variables the same name</p>

  <h2 id="ed-slide-659778-heading">Local and Global Variables</h2>
<ul>
  <li>When you create a variable inside a function, we say that the variable is “local” to that function</li>
  <li>This means that it can only be accessed by statements inside the function that created it</li>
  <li>When a variable is created outside all of your functions it is considered a “global variable”</li>
  <li>Global variables can be accessed by any statement in your program file, including by statements in any function</li>
  <li>All of the variables we have been creating so far in class (prior to discussing functions today) have been global variables</li>
</ul>
<p>You can access a global variable inside a function if you explicitly say that it is global. Look at the difference between the following two examples. However, this is a bad practice in most cases.</p>
</article>

```python
def myfunc():
  print(a) # you can access a global variable anywhere
  b=10
  print(b)

a=5
myfunc()
# this won't work; can't access local b
# print(b)
```

```python
def myfunc():
  x=10 # local variable
  print(x)
x=5 # this is a different variable than that in the function
myfunc()
print(x)
```

```python
def myfunc():
  global x
  x=10 # global variable
  print(x)
x=5 # this is the same variable as in the function
myfunc()
print(x)
```

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-long" aria-labelledby="ed-slide-659786-heading">
  <h2 id="ed-slide-659786-heading">Writing Your Own Value-Returning Functions</h2>
<ul>
  <li>You use almost the same syntax for writing a value returning function as you would for writing a normal function</li>
  <li>The only difference is that you need to include a "return" statement in your function to tell Python that you intend to return a value to the calling program, and you provide the value after the "return" keyword.</li>
  <li>The return statement causes a function to end immediately. It’s like the break statement for a loop.</li>
  <li>A function will not proceed past its return statement once encountered. Control of the program is returned back to the caller.</li>
</ul>
</article>

```python
# find the greatest common factor of two numbers x and y
def gcf(x,y):
  # the gcf can't be more than the smaller of the two numbers
  # loops from the minimum of the two numbers down to one
  for i in range(min(x,y),0,-1):
    # ends up returning the largest number that goes into both
    # evenly; that number may be one
    if ((x%i)==0) and ((y%i)==0): return i
x=gcf(7,14)
y=gcf(14,25)
z=gcf(50,75)
print(x,y,z)
print(gcf(6,9))
```

  <h2 id="ed-slide-659794-heading">Returning Multiple Results from Functions</h2>
<p>Functions can return more than one result. To do this, separate the results by commas.</p>
<p>For instance, the following function returns the lowest factor of a number and the other factor that it is paired with.</p>
</article>

```python
# a function that returns more than one number
# return the first pair of factors found of a number n
# or 1, n if  n is prime
def first_factor_pair(n):
  for i in range(2,int(n**0.5)+1):
    if ((n%i)==0):
      return i,n//i
  return 1, n

print(first_factor_pair(63))
print(first_factor_pair(27))
print(first_factor_pair(25))
print(first_factor_pair(29))
print(first_factor_pair(35))
x,y =first_factor_pair(35) # note double assignment!
print(x)
print(y)
a, b, c = 11, 12, 13
print(a,b,c)
```

  <h2 id="ed-slide-659804-heading">Seeding the Random Number Generator</h2>
<ul>
  <li>As we mentioned in an earlier class, the computer does not have the ability to generate a truly random number</li>
  <li>It uses a series of complicated mathematical formulas that are based on a known value (usually the system clock)</li>
  <li>The seed function in the random module allows you to specify the “seed” value for the pseudo-random numbers that the various random number functions generate</li>
  <li>You can seed the random number generator with whatever value you want – and you can even reproduce a pseudo-random range this way!</li>
</ul>
</article>

```python
import random
print(random.randint(1,10))

# each time you give the random module a fixed seed
# it sets it to the same initial state
# this allows you to get the same "random" results every time you run the program
# this wouldn't be possible with a truly random random number generator
# but the random module only produces "pseudo-random" numbers
random.seed(2)
print(random.randint(1,10))
print(random.randint(1,10))
print(random.randint(1,10))
for i in range(100):
  print(random.randint(1,10)," ",end="")
```
