  <h2 id="ed-slide-662236-heading">Fibonnaci Numbers</h2>
<p>The Fibonnaci sequence is the following:</p>
<p>0,1,1,2,3,5,8,13 etc.</p>
<p>It starts with 0,1 and each subsequent value is the sum of the previous two.</p>
<p>Let's write a function to calculate the nth value in the Fibonaci sequence. We pass in n and return the value.</p>
<p>You can also write a recursive version. A recursive function is a function that calls itself. It is a bit strange, but it works!</p>
<p>Note: we will not be requiring you to write recursive functions in this class, but you can if you like.</p>
<p>The output of the non-recursive version and the recursive version is the same, as expected.</p>
</article>

```python
def fibonacci(n):
  if n==1:  # first value
    return 0
  elif n==2:  # second value
    return 1
  else:
    x, y = 0, 1 # first two values
    for i in range(3,n+1):
      # another double assignment
      x, y = y, x+ y # update the last two values
    return y

for i in range(1,11):
  print(fibonacci(i),"",end="")
print()
```

```python
def fibonacci(n):
  if n==1:  # first value
    return 0
  elif n==2:  # second value
    return 1
  else:
    return fibonacci(n-1)+fibonacci(n-2) # call the function twice, recursively

for i in range(1,11):
  print(fibonacci(i),"",end="")
print()
```

  <h2 id="ed-slide-671140-heading">Understanding the Fibonacci Function</h2>
<p>Let's add some print statements to see if we can understand what is going on with the recursive Fibonacci function.</p>
</article>

```python
def fibonacci(n):
  print("Called function with",n)
  if n==1:  # first value
    rval=0
  elif n==2:  # second value
    rval=1
  else:
    rval=fibonacci(n-1)+fibonacci(n-2) # call the function twice, recursively
  print(f"Call with n={n} returned {rval}")
  return(rval)
print(fibonacci(5))
```

  <h2 id="ed-slide-662237-heading">Tax Brackets</h2>
<p>Let's say we live in a country with just two tax brackets: 15% on all income up to $100,000, and 25% on all income above. Here's a function to calculate the tax given an income.</p>
</article>

```python
def calcTax(income):
  if income<=100000:
    return 0.15*income
  else:
    return 15000+0.25*(income-100000)

print(f'${calcTax(50000):.2f}')
print(f'${calcTax(100000):.2f}')
print(f'${calcTax(150000):.2f}')
```

  <h2 id="ed-slide-662239-heading">A Factorial Function</h2>
<p>The factorial of n is the product of the first n integers. Here's a function to compute the factorial.</p>
<p>Here's a recursive version of the factorial function. It is actually easier to write the recursive version and it comes out simpler. There is no need for a loop.</p>
</article>

```python
def factorial(n):
  result=1
  for i in range(2,n+1):
    result*=i
  return(result)

for i in range(1,11):
  print(factorial(i),"",end="")
```

```python
def factorial_recursive(n):
  if n==1: return 1
  else: return n*factorial_recursive(n-1)

for i in range(1,11):
  print(factorial_recursive(i),"",end="")
```

  <h2 id="ed-slide-662241-heading">Pig Latin</h2>
<p>In Pig Latin, words that begin with a consonant move that consonant to the end and an "ay" is added. For words that begin with a vowel, "yay" is added to the end. So "tall" becomes "alltay" and "ant" becomes "antyay".</p>
<p>Let's write a function transforms a single word to Pig Latin.</p>
<p>Note that if the word contains more than one consonant at the beginning, they are supposed to all be moved to the end, that is "split" would become "itsplay". However, this would make our function harder to write, so we won't support this.</p>
</article>

```python
def pigLatin(w):
  atBeginning=True
  newWord=""
  for c in w:
    if atBeginning:
      if (c.lower() in "aeiou"):
        return w+"yay"
      else:
        ending=c+"ay"
        atBeginning=False
    else:
      newWord+=c
  return newWord+ending

print(pigLatin("Tall"))
print(pigLatin("ant"))
print(pigLatin("slant"))  # this version doesn't handle consonant clusters properly
```

  <h2 id="ed-slide-662244-heading">A Function to Draw Regular Polygons with a Specified Number of Sides, Side Length, Position and Color</h2>
<p>On replit I've created a <a class="amber-el amber-link amber-el ed-focus-outline" href="https://replit.com/@MatthewZeidenbe/PolygonFunctionTurtle#main.py" target="_blank" rel="noopener noreferrer">function</a> that allows you to specify the number of sides, side length, color and position of a regular polygon, drawn using the turtle (which is also a parameter.)</p>
<p>I've put the function in the following cell for you to paste into replit for use in the following exercise. (Don't run it in Colab; it won't work in here in Ed.)</p>
</article>

```python
def polygon(t,xstart,ystart,nSides,sideLength,color):
  # lift pen
  t.penup()
  # go to starting position
  t.goto(xstart,ystart)
  # put down pen
  t.pendown()
  # set fill color and begin fill
  t.fillcolor(color)
  t.begin_fill()
  # loop through sides
  for i in range(nSides):
   # draw a side
   t.forward(sideLength)
   #  turn left
   t.left(360/nSides)
  # end fill and lift pen
  t.end_fill()
  t.penup()
```
