  <h2 id="ed-slide-662555-heading">Dictionaries</h2>
<ul>
  <li>A dictionary in Python is like a list in that it holds indexed values</li>
  <li>Unlike a list, a dictionary doesn't use integer-based index values</li>
  <li>Instead, dictionaries use immutable objects (like Strings) to index their content</li>
  <li>In other languages dictionaries are often referred to as "associative arrays" or "hashmaps"</li>
</ul>
</article>

  <h2 id="ed-slide-662556-heading">Differences between Dictionaries and Lists</h2>
<p>Both dictionaries and lists are mutable (unlike strings)</p>
<p>But lists have an inherent order in which the objects are stored and dictionaries do not</p>
<p>List indeces are integers, but dictionary indeces can be any immutable object, like a string or an integer.</p>
<p>Dictionary indeces are called "keys."</p>
<p>Following is a sample dictionary with one integer key and two string keys. The objects indexed by the keys are called values.</p>
<p>As shown above, to create an empty dictionary, you use a pair of curly braces. We wrote:</p>
</article>

```python
names=["Amy","Raka","Alice"]
ages=[87, 22, 51]

# set up a dictionary with the same data
age={}
for i in range(len(names)):
  age[names[i]]=ages[i]

for key in sorted(age.keys()):
  print(key,age[key])

# You can also iterate over just the values in a dictionary using the values method, as in the following example:
for a in sorted(age.values()):
  print(a)

# Note that if you use the sorted function on the dictionary itself, rather than the keys or the values, it will sort the keys.
print(sorted(age))

# You can use the items method to loop over both the keys and the values, as in the following:
for key, value in age.items():
  print(key,value)
```

  <h2 id="ed-slide-662589-heading">The Dictionary get() method</h2>
<p>The get() method of a dictionary is another way to look up a key in a dictionary. Unlike indexing, the get() method will not fail if the key is not there, but will simply return the special value None.</p>
</article>

```python
names=["Amy","Raka","Alice"]
ages=[87, 22, 51]

# set up a dictionary with the same data
age={}
for i in range(len(names)):
  age[names[i]]=ages[i]

print(age.get("Amy"))
print(age.get("Tony"))
```

```python
names=["Amy","Raka","Alice"]
ages=[87, 22, 51]

# set up a dictionary with the same data
age={}
for i in range(len(names)):
  age[names[i]]=ages[i]

name=input("Enter a name:")
a=age.get(name)
if a==None:
  print("That name is not in the dictionary.")
else:
  print("The age of",name,"is",a)
```
