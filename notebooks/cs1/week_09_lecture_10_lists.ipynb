<ul>
  <li>You create a list in Python using bracket notation.</li>

  <li>You can print lists.</li>
</ul>

  <h2 id="ed-slide-662487-heading">List Mechanics</h2>
<ul>
  <li>List variables are "references"</li>
  <li>That means that they point to objects in memory, and more than one variable can point to the same object. The following creates one list and then sets it up so that two variables point to that list.</li>
</ul>
<p>This means that if you change the list using one of the variables, the change will be reflected in the other one as well.</p>
</article>

```python
mylist=[1,2,3]
mylist2=mylist
mylist2[0]=999
print(mylist)
print(mylist2)
```

  <h2 id="ed-slide-662495-heading">Drawbacks to Using "for" Loops with "in" to iterate through a list</h2>
<ul>
  <li>A "for" loop is a convenient way to iterate through a list</li>
  <li>The loop target variable assumes the value of each list item in turn</li>
  <li>However, you can't change the list this way. See the following example:</li>
</ul>
</article>

```python
mylist=[1,2,3]
for n in mylist:
  n=n*5
# no effect on the list!
print(mylist)
```

  <h2 id="ed-slide-662497-heading">Iterating Over a List Using Index Values</h2>
<p>There are two techniques for iterating over a list using index values. They are:</p>
<ul>
  <li>Setting up a counter variable outside the loop and continually updating the variable as you move to the next position in the list</li>
  <li>Using the range() function to create a custom range that represents the size of your list</li>
</ul>
</article>

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-short" aria-labelledby="ed-slide-662498-heading">
  <h2 id="ed-slide-662498-heading">Using a Counter Variable</h2>
<p>You can add a counter variable to your "for" loop to keep track of the list index. The following code doubles each element in the list.</p>
</article>

```python
mylist=[1,2,3]
counter=0
for _ in mylist:
  mylist[counter]*=2
  counter+=1
print(mylist)
```

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-short" aria-labelledby="ed-slide-662499-heading">
  <h2 id="ed-slide-662499-heading">Using the range Function</h2>
<p>The range function will also generate all of the indeces of the list, when combined with the len function. The following code is another way to double all of the elements of a list.</p>
</article>

```python
mylist=[1,2,3]
for i in range(len(mylist)):
  mylist[i]*=2
print(mylist)
```

  <h2 id="ed-slide-662503-heading">Slicing Lists</h2>

  <li>You can extract multiple items from a list.</li>

  <h2 id="ed-slide-662515-heading">Removing an Item by its Position</h2>
<p>You can also use the "del" keyword to remove an item by specifying its index position.</p>
</article>

```python
prices=[3.99,2.99,1.99]
del prices[1]
print(prices)
```

  <h2 id="ed-slide-662516-heading">Reversing a List</h2>
<p>You can use the "reverse" method on list to reverse the order of its elements.</p>
<p>This will not sort the list, but will just reverse the order.</p>
</article>

```python
names=["Alan","Betty","Dave","Chris"]
names.reverse()
print(names)
```
