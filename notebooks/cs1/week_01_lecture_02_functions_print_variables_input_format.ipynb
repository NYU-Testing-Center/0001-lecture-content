  <h2 id="ed-software-development-process-requirements-heading">The Software Development Process — Defining Requirements</h2>
<h3>Defining Requirements</h3>
<p>The software development process starts with gathering requirements. This involves meeting with the ultimate end users of the software and finding out what they need, then documenting these needs in English.</p>
<p>For instance, we might be developing a database application for theater schools for kids. These schools might need to sell tickets to shows that the schools stage with the kids, track students, track their parents, track donors (for non-profit schools), track the classes that the kids are enrolled in, track expenses and payroll, etc. All of these things are specified in the requirements document, written in English.</p>
<p>After the requirements are developed, they are passed to the programmer to implement in code.</p>
</article>

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-reading-slide  " aria-labelledby="ed-software-development-process-testing-releases-heading">
  <h2 id="ed-software-development-process-testing-releases-heading">The Software Development Process — Testing and Releases</h2>
<h3>Testing</h3>
<p>After the first version of the program is completed, it is passed to "alpha" (internal) testers.</p>
<p>As bugs are found, they are passed back to the developer for fixes. There are two main types of bugs: failure to meet the requirements, and meeting the requirements incorrectly.</p>
<p>After alpha testing, the program proceeds to public beta testing. More bugs are collected and fixed. Some programs remain in beta a long time. For instance, Gmail was in beta from 2004 to 2009.</p>
<p>After beta testing, the first official non-testing version-- the 1.0 version-- is released. Feature additions are associated with new version numbers. Major revisions are associated with full increments in the version, from 1.0 to 2.0, say. Most actively-used software is a constant work-in-progress.</p>
</article>

<p>What if we want to swap the values of two variables? We need a third variable as a temporary holder of one of the values.</p>

```python
v1=10
v2=20
print(v1,v2)
tempvar=v1
v1=v2
v2=tempvar
# remove the temporary variable
del tempvar
print(v1,v2)
```

  <h2 id="ed-format-function-strings-heading">The Format Function: Strings</h2>
<p>The format function formats strings and numbers according to a "format string." Let's look at some examples, first with strings, then with integers, and then with floating point numbers:</p>
</article>

```python
print("123456789-123456789-123456789-123456789-")
# two strings + a third string
# left justified in a field of width 20, then right justified in a field of width 10
# < means left-justified, and > right-justified
print(format("Hello","<20s"),format("there",">10s")," world",sep="")
```

</article>

```python
# integers
print("123456789-123456789-123456789-123456789-")
# left justified, then right justified
print(format(23,"<20d"),format(192,">10d"), sep='')
```
