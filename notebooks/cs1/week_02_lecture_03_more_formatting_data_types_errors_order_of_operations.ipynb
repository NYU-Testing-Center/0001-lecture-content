  <h2 id="ed-slide-652873-heading">Formatting with the format function, again</h2>
<p>"&gt;" and "&lt;" right and left justify, respectively. The format function returns a string which can be printed. It takes two arguments, the object to be formatted and the formatting string. The string returned is the length of the width of the field in the formatting string.</p>
<p>The "s" format specifier is for formatting strings. Here's an example that starts by formatting two strings of length five. I use a "ruler" which helps us see how things come out.</p>
</article>

```python
ruler="123456789-"
h="hello"
t="there"
# format takes two arguments, the object to be formatted and the formatting string
s1=format(h,"<10s") # returns a string of length 10
s2=format(t,">7s") # returns
print(s1)
print(s2)
# the len() function returns the length of a string
print(len(h),len(t),len(s1),len(s2))
# "\n" advances to the next line
print(ruler*2,"\n",s1,s2,sep="")
```

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-short" aria-labelledby="ed-slide-652874-heading">
  <h2 id="ed-slide-652874-heading">Formatting Integers</h2>
<p>Integers are formatted similarly to strings, but instead of the "s" format specifier, use the "d" format specifier, plus a width, as in the following example which prints out two integers, one with a width of 10 and another with a width of 7.</p>
</article>

```python
ruler="123456789-"
print(ruler*2,"\n",format(43213,"<10d"),
      format(12349,">7d"),sep='') # this is one way to continue a line
```

<article class="nyu-lecture-slide nyu-ed-slide lecture-shell nyu-explanation-code-slide  nyu-code-density-long" aria-labelledby="ed-slide-652875-heading">
  <h2 id="ed-slide-652875-heading">Formatting floating point numbers</h2>
<p>Floating point numbers use the "f" format specifier. For instance, "10f.2" means the number should be formatted in a string of a total length of 10 with 2 places after the decimal point. The string is padded as needed on the left by spaces.</p>
<p>Note that format does not actually change the underlying floating point number it is formatting. This is done by the round function if desired. But it is not necessary to change a floating point number in order to show it rounded during printing by using the format function.</p>
<p>Here is an example.</p>
</article>

```python
a=1/6
print(a)
# ask the user for the two components of the format specifier
w=input("What width would you like?")
p=input("How many places after the decimal point?")
# construct the specifier from the numbers input
formatspec=w+"."+p+"f"
print(formatspec)
# format using the specifier
b=format(a,formatspec)
print(b)
b=format(a,"10.4f")
print(b)
# formatting doesn't change the underlying value
# note a is unchanged; this formatting does not round the variable a itself
print(a)
# to change the underlying value, use the round function
a=round(a,3) # this "round" function actually rounds the variable a
print(a)
print(round(a,1))
print(a)
```

  <h2 id="ed-slide-652876-heading">Formatting Numbers Using Commas</h2>
<p>The "f" and "d" format specifiers allow for formatting numbers so that commas appear every three digits starting from the right of the number, as in the following example.</p>
</article>

```python
ruler="123456789-"
a=12345.5789
print(format(a,".3f")) # without comma separators
print(format(a,",.3f")) # with comma separators
print(format(a,",.0f")) # comma separators and rounding to nearest whole number
print(2*ruler)
# with comma separators, field width and justificaiton
print(format(a,">15,.1f"))
print(format(a,"<10,.2f"),format(a,">10,.1f"), sep="")
x=123455555555555555555929292933999129394995919293293932991233
print(x)
print(format(x,",d"))
```

  <h2 id="ed-slide-652900-heading">Format Specification Mini-Language</h2>
<p>The rules for specifying format are given by Python's format specification mini language, which is described <a class="amber-el amber-link amber-el ed-focus-outline" href="https://www.google.com/url?q=https%3A%2F%2Fdocs.python.org%2F3%2Flibrary%2Fstring.html%23formatspec" target="_blank" rel="noopener noreferrer">here</a>. There are more formatting types and other syntax that we haven't covered.</p>
<p>In addition, one of the other ways to control format through the use of "<a class="amber-el amber-link amber-el ed-focus-outline" href="https://docs.python.org/3/tutorial/inputoutput.html#tut-f-strings" target="_blank" rel="noopener noreferrer">f-strings</a>" (also known as "formatted string literals") which have become popular more recently. We won't cover these in detail, but the following are a couple of examples. You can use f-strings in your programs if you like. They are my favorite way to do formatting in Python because they are so concise compared to the other methods. They use the same format specifiers as the other methods.</p>
<p>Here is an example:</p>
</article>

```python
# variables are included in the f-string in braces
# format is given after a colon in the braces
price=2.5
# version using an f-string
print(f'The price is ${price:.2f}')
# version using the format statement
print("The price is $",format(price,".2f"),sep="")
speed_of_light=186282.39705122
print(f'The speed of light is {speed_of_light:,.3f} miles per second.')
```
