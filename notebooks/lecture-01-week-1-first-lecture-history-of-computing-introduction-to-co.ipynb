## Alan Turing and the Turing Machine

#### British mathematician Alan Turing (1912-1954) developed many of the ideas that led to the development of the modern computer

#### One of his main conceptions was the Turing Machine, which consisted of a “head” which reads and writes symbols from/to the tape, which moves backwards and forwards so that different parts of the tape. Turing proved that the Turing Machine could execute a very wide range of programs.

#### Turing influenced John Von Neumann, one of the leading early computer designers.

#### Turing had a tragic life because he was treated horribly by the British government because he was gay, despite his invaluable services in code-cracking during WWII

## The Processor (Central Processing Unit or CPU)

- #### Every computer has at least one processor and most have several processors
- #### The processor acts as the computer’s “brain” and coordinates all its operations.
- #### A processor can perform four distinct operations (typically within one or more nanoseconds):
  - #### Receive a new instruction (fetch)
  - #### Make sense of this instruction (decode)
  - #### Perform the action defined by this instruction (execute)
  - #### Store the result of the action (store)

#### A processor is made up of miniaturized transistors and other components and circuits on a semiconductor “chip.” The following shows a microprocessor chip next to a penny.

<div style="text-align:center;margin:0.4em 0;"><img src="images/XI5dcKJ726BBOuOjtugVyuBV.png" alt="" style="max-width:100%;width:450px;height:auto;"></div>

## Moore's Law

- #### Advances in microchip design have facilitated the development of computer technology thanks to Moore’s Law. Gordon Moore, one of the founders of Intel, first formulated the law.
- #### Moore’s Law is the empirical observation that the complexity of an integrated circuit (microchip) doubles about every two years.
- #### It’s unclear how much longer Moore’s Law will continue to describe the state of affairs, but it has done so in the past.
- #### Moore’s law is why computers have gotten faster and faster over time.

## Computer Memory (RAM)

- #### Computers, like us, need short-term memory in order to function properly
- #### We call this “Random Access Memory” or RAM and it is very fast!
- #### We call it “random” since the computer can selectively read and write to any place in its RAM in a fixed (short) amount of time
- #### When a computer turns off, it loses what is stored in RAM (we call this volatility)
- #### In general, the more RAM a system has, the better it performs. Often the amount of RAM is more important to performance than CPU speed, although that depends on the application.

## Long Term Computer Storage

- #### Temporary storage is accomplished in memory (RAM)
- #### We use RAM for short term memory because it is very fast, but it is also volatile and loses it contents when the computer is shut off
- #### Long term storage is, by definition, non-volatile- it has the ability to persist even if the power is off
- #### It also typically holds far more data than the RAM can
- #### Long term storage devices include hard drives, solid-state drives, and USB “key” drive
- #### Sometimes this storage is available in the cloud

## Computers Only Understand Binary Codes

- #### The underlying language, the machine language, of a computer is numerical and is expressed in binary (zeros and ones).
- #### Each code represents an instruction to the CPU (e.g. add two numbers)
- #### High-level languages, like Python and Java, are translated into machine language by a program called an interpreter or a compiler
- #### Python is usually an interpreted language but compilers exist
- #### Compilers produce faster code than interpreters

The program below converts temperature from Fahrenheit to Celsius.

The program uses the input function to get a value of Fahrenheit degrees from the user. It then converts that input to a floating point number using the float function. It stores the result in the degreesF variable. It then converts degreesF to degreesC using a formula, and prints out the result.

```python
degreesF=float(input("How many degrees Fahrenheit?"))
degreesC=(degreesF-32)*5/9
print("That is", degreesC,"degrees Celsius")
```
