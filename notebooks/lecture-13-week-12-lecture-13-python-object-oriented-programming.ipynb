## Object-Oriented Programming with Programmer-Defined Objects and Classes in Python

Everything in Python is an object, and every variable in Python points to an object which is an instance of a class.

You can think of the class as a set and objects in that class as members of that set. For instance, a and b are both objects that are instances of the class "string."

```python
# All objects have methods that are specific to that type
s="the rain in Spain"
print(s.lower())
x=1.5
try:
  print(x.lower())
except:
  print("can't apply the 'lower' method to a float")
#print(x.lower())
```

```python
a="hello"
print(type(a)) # str
b="world"
print(type(b)) # str
y=2.6
print(type(y)) # float
z=5
print(type(z)) # int
def addOne(x):
  return x+1
z=addOne(z)
print(z)
print(type(addOne)) # function
```

c and d are both instances of the class "list."

```python
c=["red","green","blue"]
print(c)
print(type(c)) # list
d=c+['purple']
print(d)
print(type(d)) # list
d2={1:1, 2:4, 3:9, 4:16}
print(d2)
print(type(d2)) # dict
```

## Objects and their Classes have Associated Methods

Methods are functions that are specific to objects in a class. For instance, reverse and append are methods that work on lists. Methods are accessed using objects and "dot notation."

```python
c=["red","green","blue"]
d=c+['purple']

#x=1
#x.reverse
print(d)
d.reverse() # destructively reversing (replaces the original )
print(d)
d.append("orange")
print(d)
div5=list(range(5,101,5))
print(div5)
div5.reverse()
print(div5)
```

Keys, values and items are methods that work on dictionaries.

```python
heights={"Sengbeng":64, "Mary":65, "Ganesh":69}
print(heights.keys())
print(heights.values())
print(heights.items())
import string
print(string.ascii_lowercase)
l=list(string.ascii_lowercase)
print(l)
try:
  print(l.items())
except:
  print("can't run the items method on a list")
#print(l.items())
```

```python
x=["abc","bcd"]
x.reverse()
print(x)
```

## Python Programmer-Defined Objects and Classes

#### Terminology

- A class defines a type of object
- An object is an instance of a class
- When defining a class, we define attributes and methods of that class
- An attribute is a variable associated with a class and its objects
- A method is a function associated with a class and its objects

#### We can define our own classes and then create objects using them.

#### Syntax of Programmer-Defined Classes

- "Class" keyword defines a class giving the class name
- Methods are defined as functions indented under the class
- "Self" refers to the object itself
- All methods refer to self and optionally some other arguments
- Special "dunder" or magic methods begin and end with double underscore

## Class methods

- \_\_init\_\_ method is used to create an object
- \_\_str\_\_ method is used to tell the print function how to print an object
- \_\_eq\_\_ tells the == operator when two objects are equal
- Methods are called with dot notation
- Objects are created with the class name and the arguments to \_\_init\_\_

A class for dogs with four methods:

```python
# define a class
class Dog:
  def __init__(self,name,weight,age,got_rabies_shot):
    self.name=name
    self.weight=weight
    self.age=age
    self.got_rabies_shot=got_rabies_shot
  def __str__(self):
    s=self.name+" is a dog that weighs "
    s+=str(self.weight)+" pounds and is "
    s+=str(self.age)+" years old."
    if self.got_rabies_shot:
      s+="\nVaccinated for Rabies."
    else:
      s+="\nNot vaccinated for Rabies."
    return s
  def have_birthday(self):
    self.age+=1
  def compare_age(self, other):
    if self.age>other.age:
      print(self.name,"is older.")
    elif self.age<other.age:
      print(other.name,"is older.")
    else:
      print(self.name,'and',other.name,'are the same age.')

# create an instance (calls __init__)
dog1=Dog("Fido",25,1,False)
# Dog.__init__(dog1,"Fido",25,1,False)
# print out the attributes
print(dog1.name)
print(dog1.weight)
print(dog1.age)
print(dog1.got_rabies_shot)
print(dog1)
dog2=Dog("Spot",33,5,False)
print(dog2)
dog2.compare_age(dog1)
Dog.compare_age(dog2,dog1)
dog1.compare_age(dog2)
Dog.compare_age(dog1,dog2)


# change dog1's name
dog1.name="Rover"
# print the object (calls the __str__ method)
print(dog1)
# if he has a birthday he will be one year older
dog1.have_birthday()
print(dog1)
```

## A Point Class

The following is a class that defines points in the X-Y plane.

It has four methods:

- A "dunder" \_\_init\_\_ method which tells how a point is initialized (with x and y values)
- A "dunder" \_\_str\_\_ method which tells how a point is printed
- A "dist\_to\_origin" method which tells how far a point is from the origin.
- A "manhattan\_distance" method which tells the "Manhattan Distance" between two points, which is the distance between them using only horizontal and vertical motion in the x-y grid.

It also has a "doc string" which is the string right below the class definition.

```python
class point:
  '''Defines a point in terms of its x and y coordinates'''
  def __init__(self, x, y):
    self.x=x
    self.y=y
  def __str__(self):
    return f"({self.x}, {self.y})"
  def dist_to_origin(self):
    return (self.x**2+self.y**2)**(0.5)
  def manhattan_distance(self,other):
    return abs(self.x-other.x)+abs(self.y-other.y)

p1=point(3,2)
print(p1)
p2=point(1,1)
print(p2)
print(p2.dist_to_origin())
print(p1.manhattan_distance(p2))
print(p2.manhattan_distance(p1))
print(point.__doc__)
```

## A Square Class

Following is a a class that defines a square in terms of its lower left corner x, y coordinates and the length of its side.

It has the \_\_init\_\_ method and one additional method, "upper\_right", which finds the point at the upper right of the square.

It also has a doc string.

```python
class point:
  '''Defines a point in terms of its x and y coordinates'''
  def __init__(self, x, y):
    self.x=x
    self.y=y
  def __str__(self):
    return f"({self.x}, {self.y})"
  def dist_to_origin(self):
    return (self.x**2+self.y**2)**(0.5)
  def manhattan_distance(self,other):
    return abs(self.x-other.x)+abs(self.y-other.y)

class square:
  '''Defines a square in terms of the x and y
  coordinates of its lower left corner and the
  length of a side'''
  def __init__(self,llx,lly,side):
    self.ll=point(llx,lly)
    self.side=side
  def upper_right(self):
    return point(self.ll.x+self.side,self.ll.y+self.side)


sq1=square(0,0,2)
print(sq1.ll)
print(sq1.upper_right())
print(square.__doc__)
```

## A Random Playing Card Class

Following is a class that deals a random card from a deck (with replacement, which means you can get the same card twice.)

It has a \_\_str\_\_ method for printing, and methods which tell whether two cards have the same suit or are the same card. The latter is the \_\_eq\_\_ method which defines the "==" operator for the class.

```python
class random_card:

  def __init__(self):
    suits=("Hearts","Diamonds","Spades","Clubs")
    values=("Ace","2","3","4","5","6","7","8","9","10","Jack","Queen","King")
    import random
    self.suit=random.choice(suits)
    self.value=random.choice(values)
  def __str__(self):
    return self.value+" of "+self.suit
  def same_suit(self,other):
    return self.suit==other.suit
  def __eq__(self,other):
    return (self.suit==other.suit) and (self.value==other.value)

card1=random_card()
print(card1)
card2=random_card()
print(card2)
print(card2.same_suit(card1))
print(card1.same_suit(card2))

# Let's see how long it takes to get two cards that are identical.
def findMatchingPair():
  i=0
  while True:
    i+=1
    card1=random_card()
    card2=random_card()
    if card1==card2: return i
print(f"It took {findMatchingPair()} pairs of cards to get the same card twice.")

total=0
for i in range(1000):
  total+=findMatchingPair()
average=total/1000
print(average)
```
