## Formatting with the format function, again

">" and "<" right and left justify, respectively. The format function returns a string which can be printed. It takes two arguments, the object to be formatted and the formatting string. The string returned is the length of the width of the field in the formatting string.

The "s" format specifier is for formatting strings. Here's an example that starts by formatting two strings of length five. I use a "ruler" which helps us see how things come out.

````python
ruler="123456789-"
h="hello"
t="there"
# format takes two arguments, the object to be formatted and the formatting string
s1=format(h,"<10s") # returns a string of length 10
s2=format(t,">7s") # returns
print(s1)
print(s2)
# the len() function returns the length of a string
print(len(h),len(t),len(s1),len(s2))
# "\n" advances to the next line
print(ruler*2,"\n",s1,s2,sep="")
````

## Formatting Integers

Integers are formatted similarly to strings, but instead of the "s" format specifier, use the "d" format specifier, plus a width, as in the following example which prints out two integers, one with a width of 10 and another with a width of 7.

````python
ruler="123456789-"
print(ruler*2,"\n",format(43213,"<10d"),
      format(12349,">7d"),sep='') # this is one way to continue a line
````

## Formatting floating point numbers

Floating point numbers use the "f" format specifier. For instance, "10f.2" means the number should be formatted in a string of a total length of 10 with 2 places after the decimal point. The string is padded as needed on the left by spaces.

Note that format does not actually change the underlying floating point number it is formatting. This is done by the round function if desired. But it is not necessary to change a floating point number in order to show it rounded during printing by using the format function.

Here is an example.

````python
a=1/6
print(a)
# ask the user for the two components of the format specifier
w=input("What width would you like?")
p=input("How many places after the decimal point?")
# construct the specifier from the numbers input
formatspec=w+"."+p+"f"
print(formatspec)
# format using the specifier
b=format(a,formatspec)
print(b)
b=format(a,"10.4f")
print(b)
# formatting doesn't change the underlying value
# note a is unchanged; this formatting does not round the variable a itself
print(a)
# to change the underlying value, use the round function
a=round(a,3) # this "round" function actually rounds the variable a
print(a)
print(round(a,1))
print(a)
````

## Formatting Numbers Using Commas

The "f" and "d" format specifiers allow for formatting numbers so that commas appear every three digits starting from the right of the number, as in the following example.

````python
ruler="123456789-"
a=12345.5789
print(format(a,".3f")) # without comma separators
print(format(a,",.3f")) # with comma separators
print(format(a,",.0f")) # comma separators and rounding to nearest whole number
print(2*ruler)
# with comma separators, field width and justificaiton
print(format(a,">15,.1f"))
print(format(a,"<10,.2f"),format(a,">10,.1f"), sep="")
x=123455555555555555555929292933999129394995919293293932991233
print(x)
print(format(x,",d"))
````

## Format Specification Mini-Language

The rules for specifying format are given by Python's format specification mini language, which is described [here](https://www.google.com/url?q=https%3A%2F%2Fdocs.python.org%2F3%2Flibrary%2Fstring.html%23formatspec). There are more formatting types and other syntax that we haven't covered.

In addition, one of the other ways to control format through the use of "[f-strings](https://docs.python.org/3/tutorial/inputoutput.html#tut-f-strings)" (also known as "formatted string literals") which have become popular more recently. We won't cover these in detail, but the following are a couple of examples. You can use f-strings in your programs if you like. They are my favorite way to do formatting in Python because they are so concise compared to the other methods. They use the same format specifiers as the other methods.

Here is an example:

````python
# variables are included in the f-string in braces
# format is given after a colon in the braces
price=2.5
# version using an f-string
print(f'The price is ${price:.2f}')
# version using the format statement
print("The price is $",format(price,".2f"),sep="")
speed_of_light=186282.39705122
print(f'The speed of light is {speed_of_light:,.3f} miles per second.')
````

## Data Types

Python stores different types of data as objects in the RAM (random access memory) of the computer.

Python supports many types of objects , but initially we will work with four types:

- strings (e.g. "hello") (class 'str')
- integers (e.g. 17, no decimal point) (class 'int')
- floating point numbers (e.g. 5.2, decimal point) (class 'float')
- boolean values (True or False) (class 'bool')

Let's generate a random integer between 1 and 100 and then see if it is even or odd. Each time I run the program I will get a different number generated by the "random" module.

````python
import random
my_random_number=random.randint(1,100)
s1="The random number is:"
print(s1, my_random_number)

s2="Is it even?"
if ((my_random_number%2)==0):
  even=True
else:
  even=False
print(s2,even)
print(type(even)) # the "type" function tells us the type of an object
````

## Computing the Area and Perimeter of a Circle

We can use pi to compute the area and perimeter of a circle given its radius, as in the following example:

````python
import math
radius=float(input("Enter the radius of the circle:"))
perimeter=2*math.pi*radius
area=math.pi*(radius**2)
print("The perimeter of the circle is:", format(perimeter,"10.3f"))
print("The area of the circle is", format(area,"10.3f"))

# both of the new variables are floats
print(type(area))
print(type(perimeter))
````

## Python: Dynamically Typed, Not Statically Typed

- You don't need to declare the type of data that your variables will hold (which you do need to do in many other languages)
- In addition, the type of data held by a variable can change and is defined by what you put in it.
- This is called "dynamic" typing" (as opposed to "static" typing)

For example:

````python
my_var="Hello"
print(my_var,type(my_var))
my_var=3.0
print(my_var,type(my_var))
my_var=3
print(my_var,type(my_var))
````

## Errors in Python

As you write code, you are likely to encounter errors. These errors will be one of the following three types:

1. Syntax errors: your program isn't correct Python
2. Runtime errors: your program is correct Python, but something averse happens when you run it, like dividing by zero
3. Logic errors: your program is correct Python, and there aren't any runtime errors, but you made a logical error when you wrote your program

## An Example of a Syntax Error

The following is a syntax error, because the string delimiters don't match.

````python
print("hello world')
````

## A Runtime Error

The following code will generate a runtime error if you don't enter a number, because the conversion to float will fail.

````python
num=float(input("Please enter a number:"))
print(num)
````

## A Logic Error

The following code represents a logic error because you subtracted when you intended to add.

````python
num1=float(input("Enter a number:"))
num2=float(input("Enter another number:"))
print("The sum is",num1-num2)
````

## How to Debug

- Write your program in small pieces and test each piece to make sure it does what you think it does.
- Stop frequently and test often.
- Take a coffee break after you finish each piece successfully.
- Comment your program a lot so that you remember what you did, and so others can understand it.
  - Assignments will be marked down for lack of comments
- Insert print statements for debugging and "comment them out" when desired (you can delete them in your final code)

## Python's Order of Operations

Python uses the standard order of operations (PEMDAS)

## Examples of the Order of Operations

Here are some examples of the order of operations:

````python
print(4*3**2) # 36
print(4*(3**2)) # 36
print((4*3)**2) #144

print(4/2*6) # 12.0
print(4/(2*6)) # 0.3333333333333333

print(4-2+6) # 8
print(4-(2+6)) # -4
````

## Converting math formulas to Python

Multiplication and division are written differently in math and in Python and other computer languages.

For instance, Einstein's famous formula:

$$
E=mc^2
$$

is written as follows in Python.

````python
m=10
c=186282
E=m*(c**2)
print(E)
````

## Another formula

The formula

$$
\frac{1 }{x+2}
$$

is written as follows in Python.

````python
x=8
y=1/(x+2)
print(y)
````

## The Modulo Operator

In Python, the module operator "%" as in "a%b" returns the remainder when a is divided by b; a and b must be integers.

For example:

````python
print(15%4)
print(15%5)
````

## Type of a Calculation

If all of the values in a calculation are int, the result will sometimes be an int and sometimes a float.

````python
x=5*3+4*7
print(x, type(x))

x=5/3
print(x, type(x))
x=6/3
print(x,type(x))
x=int(6/3)
print(x,type(x))
````

## Type of a Calculation (2)

If at least one of the values is a float, the result will be a float.

````python
x=2.0+1
print(x, type(x))
````

## Line Continuation

A backslash can be used to continue a line in Python if you want to split it into multiple lines, for instance, if the line gets too long.

````python
x=5*3 + \
  4*7
print(x)
````

## Line Continuation (2)

Anything in parentheses can be continued onto multiple lines:

````python
x=(5*3+
   4*7)
print(x)

x=3
y=4
z=5
print(x,
      y, z)
````

## String Concatenation

The "+" operator in the context of strings means concatenation, that is, adding one string to the end of another.

````python
s1="hello"
s2="world"
s3=s1+" "+s2
print(s3)
print(len(s3))
````

## String Repetition

The "\*" operator in the context of strings allows you to repeat a string.

````python
y="Yadda "*10
print(y)
print(len(y))
print("-"*40)
````
