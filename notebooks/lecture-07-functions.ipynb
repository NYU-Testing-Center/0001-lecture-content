## Python Functions

- A function is a group of statements that exist within a program for the purpose of performing a specific task
- We have already been using a number of Python's built in functions, including:
  - range(): generates a range of integers
  - print(): outputs data to the console
  - len(): measures the length of an object, like a list or a string
  - random.randint(): generates a random integer within a range

## Defining Functions

- Functions, like variables, must be named and created before you can use them
- The same naming rules apply to both variables and functions:
  - You can't use any of the keywords Python reserves to itself, like "if", "while", or "for"
  - No spaces in a name
  - First character must be in the alphabet, or be the underscore character
  - After the first character, you can use alphabetic or numeric digits or the underscore
  - Python is case-sensitive, which means, for instance, that "A" and "a" are not the same

The following is an example function definition, "myfunction". Note we use the def keyword to define the function; to call it we refer to it by its name.

````python
# define a function
def myfunction():
  print("Hello, I'm in the function!")

# call it twice with an intervening print
myfunction()
print("Hi, I'm in the main program.")
myfunction()
````

## Some Notes on Functions

- To run a function, you "call" it. This transfers control up to the function and the statements indented under the function are run.
- Once a function has completed, Python continues with the line after the call.
- Functions must be defined before they are called. We generally place all of our functions at the beginning of our programs.

The following is an example that calls two functions.

````python
# note hello doesn't need to be defined first as long as it is defined before it is called.
def goodbye():
  print("See ya!")
def hello():
  print("Hello there!")

hello()
goodbye()
````

The following is an example of a function that calls another function. This is allowed, and is actually common.

````python
def main():
  print("I have a message for you.")
  # function calls other function "message"
  message()
  print("Goodbye!")
def message():
  print("The password is 'Open Sesame'.")
main()
````

## Local Variables

Local variables are variables that exist only inside the function (inside its "scope," which is the code block associated with it) and do not exist outside of it. This is best seen with an example.

````python
def bugs():
  numbugs=int(input("How many bugs?"))
  print(f"You entered {numbugs} bugs.")

bugs()
# if you try to print the variable numbugs here, it won't work
# since it is defined only in the function and is local to it
#print(numbugs)
````

## Something Weird

The following is a bit weird. There are two variables have the same name, but they are distinct, different variables. One exists inside the function and is local to it, whereas the other is in the main program.

````python
def bugs():
  numbugs=int(input("How many bugs?")) # local version
  print(f"You entered {numbugs} bugs.")

numbugs="Forty-one" # global version
bugs()
# now you can print it, but it will be the version from the main program.
# there are two variables with the same name
# one is local, and the other is global
print(numbugs)
````

## Something Weirder

Here is something even weirder. We have three versions of the variable, all with the same name, but all different.

````python
def bugs():
  numbugs=int(input("How many bugs?")) # local version
  print(f"You entered {numbugs} bugs.")
  bugs2()
def bugs2():
  numbugs=int(input("What is the bug count?")) # another local version
  print(f"You told me there were {numbugs} bugs.")

numbugs=4.0 # global version
bugs()
print(numbugs)
````

## Don't Re-use the Same Variable Name

Don't confuse yourself and others by giving different variables the same name

While it is technically possible to have a situation as in the prior example, three variables, two local, one global, all distinct, but with the same name, don't do it! You will only confuse yourself and others. Give every variable in your program a different name, and use different names in each function. There are some exceptions to this rule, but it is generally a good one.

## Local and Global Variables

- When you create a variable inside a function, we say that the variable is “local” to that function
- This means that it can only be accessed by statements inside the function that created it
- When a variable is created outside all of your functions it is considered a “global variable”
- Global variables can be accessed by any statement in your program file, including by statements in any function
- All of the variables we have been creating so far in class (prior to discussing functions today) have been global variables

````python
def myfunc():
  print(a) # you can access a global variable anywhere
  b=10
  print(b)

a=5
myfunc()
# this won't work; can't access local b
# print(b)
````

You can access a global variable inside a function if you explicitly say that it is global. Look at the difference between the following two examples. However, this is a bad practice in most cases.

````python
def myfunc():
  x=10 # local variable
  print(x)
x=5 # this is a different variable than that in the function
myfunc()
print(x)
````

````python
def myfunc():
  global x
  x=10 # global variable
  print(x)
x=5 # this is the same variable as in the function
myfunc()
print(x)
````

## Issues with Global Variables

- Global variables can make debugging difficult
- Functions that use global variables are generally dependent on those variables, making your functions less reusable in other programs
- However, there are some situations in which using global variables may make sense, although some people think they should never be used.

## One-Way Communication

Data that is passed into a function as a parameter may be modified by the function, but since the parameter is local to the function, this is not felt outside the function. See the following:

````python
def test(x):
  x+=1 # x is incremented
  print(x) # and you can see it

y=5
# the 5 goes into the function
test(y)
# but it doesn't come back out
# communication is in only one direction
print(y) # but the glabel y is not affected
````

## Value-Returning Functions

- Value returning functions are functions that return a value to the part of the program that initiated the function call
- They are almost identical to the type of functions we have been writing so far, but they have the added ability to send back information at the end of the function call
- We have been using these all semester, as in the following:

````python
# the input function returns a string
name=input("What is your name?")
print(f"Hello, {name}!")
````

````python
# the following function returns a random integer in the specified range
import random
n=random.randint(1,5)
print(f"The random number is {n}")
````

## Writing Your Own Value-Returning Functions

- You use almost the same syntax for writing a value returning function as you would for writing a normal function
- The only difference is that you need to include a "return" statement in your function to tell Python that you intend to return a value to the calling program, and you provide the value after the "return" keyword.
- The return statement causes a function to end immediately. It’s like the break statement for a loop.
- A function will not proceed past its return statement once encountered. Control of the program is returned back to the caller.

````python
# find the greatest common factor of two numbers x and y
def gcf(x,y):
  # the gcf can't be more than the smaller of the two numbers
  # loops from the minimum of the two numbers down to one
  for i in range(min(x,y),0,-1):
    # ends up returning the largest number that goes into both
    # evenly; that number may be one
    if ((x%i)==0) and ((y%i)==0): return i
x=gcf(7,14)
y=gcf(14,25)
z=gcf(50,75)
print(x,y,z)
print(gcf(6,9))
````

## Don't Confuse Printing and Returning

Printing a value inside a function is not the same as returning it. If you only print, nothing will be returned.

````python
def double(x):
  print(2*x)

y=double(4)  # nothing is returned here, defined as "None"
print(y)
````

````python
def double(x):
  return 2*x

y=double(4) # now something is returned
print(y)
````

## Boolean-Returning Functions

Functions can return any type of value.

So you can have Boolean functions, which return a Boolean value, like the following one:

````python
def Disney(name):
  # converts name to lower case and sees if it is in the list
  # note the "in" operator returns a Boolean result
  return name.lower() in ["mickey","minnie","donald","daisy","goofy","pluto","pete"]

print(Disney("Pluto"))
print(Disney("Bugs"))
````

## Returning Multiple Results from Functions

Functions can return more than one result. To do this, separate the results by commas.

For instance, the following function returns the lowest factor of a number and the other factor that it is paired with.

````python
# a function that returns more than one number
# return the first pair of factors found of a number n
# or 1, n if  n is prime
def first_factor_pair(n):
  for i in range(2,int(n**0.5)+1):
    if ((n%i)==0):
      return i,n//i
  return 1, n

print(first_factor_pair(63))
print(first_factor_pair(27))
print(first_factor_pair(25))
print(first_factor_pair(29))
print(first_factor_pair(35))
x,y =first_factor_pair(35) # note double assignment!
print(x)
print(y)
a, b, c = 11, 12, 13
print(a,b,c)
````

## Modules

- All programming languages come pre-packaged with a standard library of functions that are designed to make your job as a programmer easier
- Some of these functions are built right into the “core” of Python (print, input, range, etc)
- Other more specialized functions are stored in a series of files called “modules” that Python can access upon request by using the “import” statement
  - import random
  - import time

````python
# after you import a module, you can get help on it
# but it is easier to look online
import time
print(time.gmtime()) # Time in London (Greenwich Mean Time)
#help(time)
````

## Modules as "Black Boxes"

- The import statement tells Python to load the functions that exist within a specific module into memory and make them available in your code
- Because you don’t see the inner workings of a function inside a module we sometimes call them [“black boxes”](https://www.google.com/url?q=https%3A%2F%2Fwww.elderresearch.com%2Fblog%2Fhow-and-why-to-interpret-black-box-models%2F)
- A "black box" describes a mechanism that accepts input, performs an operation that can’t be seen using that input, and produces some kind of output

## Functions in Modules

We call functions that exist within a module by using “dot notation” to tell Python to run a function that exists in that module, as in the following example:

````python
import random
# following function takes a list and gives you a random item in it
print(random.choice(["Rock","Paper","Scissors"]))
````

## Seeding the Random Number Generator

- As we mentioned in an earlier class, the computer does not have the ability to generate a truly random number
- It uses a series of complicated mathematical formulas that are based on a known value (usually the system clock)
- The seed function in the random module allows you to specify the “seed” value for the pseudo-random numbers that the various random number functions generate
- You can seed the random number generator with whatever value you want – and you can even reproduce a pseudo-random range this way!

````python
import random
print(random.randint(1,10))

# each time you give the random module a fixed seed
# it sets it to the same initial state
# this allows you to get the same "random" results every time you run the program
# this wouldn't be possible with a truly random random number generator
# but the random module only produces "pseudo-random" numbers
random.seed(2)
print(random.randint(1,10))
print(random.randint(1,10))
print(random.randint(1,10))
for i in range(100):
  print(random.randint(1,10)," ",end="")
````
