## Variables So Far

So far we have been working with variables that can hold one piece of data, like floats, integers, strings, and booleans.

```python
x=15.5 # float
y=10 # integer
z="Hello There" # string
w=True #Boolean
print(x,y,z,w)
```

## Lists

There are also objects in Python that can hold multiple items of data. These include lists, tuples, dictionaries, and sets. This week we will talk about lists.

As you might imagine, a list is an object that can hold a list of items.

A list is a kind of sequence object in Python because the items in the list have an order.

## Creating a List in Python

- You create a list in Python using bracket notation.
- The code below will create a list in Python that holds three integers-- 1, 2, and 3-- in that order
- You can think of a list as a "book" that holds a series of sheets of paper, each of holds one item.
- You can print lists.

```python
my_list=[1,2,3]
print(my_list)
```

Lists can be composed of a single data type or a mix of data types.

```python
l1=["Adam","Eve","Cain","Abel"]
l2=["Joe",True,4.1,5]
print(l1)
print(l2)
```

You can use the repetition operator ("\*") to have Python repeat a list, much as we have done with strings.

```python
# this will create a list consisting of 12 strings.
song=["Tra","La","La"]*4
print(song)
```

You can use the concatenation ("+") to combine lists, much as we have done with strings.

```python
# here a line of a song is represented by a list of the words in that line.
line1=["Ob", "la", "di"]
line2=["Ob", "la", "da"]
combinedLines=line1+line2
print(combinedLines)
```

## Indexing into Lists

- You can index into lists in a similar manner that you did with strings. In the case of strings, indeces pull out individual characters; in the case of lists they pull out individual list elements.
- The first index into a list is zero.

```python
line1=["Ob", "la", "di"]
line2=["Ob", "la", "da"]
combinedLines=line1+line2

print(combinedLines)
print(combinedLines[0])
print(combinedLines[2])
print(combinedLines[-1])
```

## Lists are Mutable

Lists, unlike strings, are mutable. That means that their elements can be changed.

```python
# create a list
l=[1,2,3]
print(l)
# change its third element
l[2]=99
print(l)
```

## List Mechanics

- List variables are "references"
- That means that they point to objects in memory, and more than one variable can point to the same object. The following creates one list and then sets it up so that two variables point to that list.

```python
mylist=[1,2,3]
mylist2=mylist
print(mylist)
print(mylist2)
```

This means that if you change the list using one of the variables, the change will be reflected in the other one as well.

```python
mylist=[1,2,3]
mylist2=mylist
mylist2[0]=999
print(mylist)
print(mylist2)
```

## Copying a List

- Python will only create new lists when you use the square bracket list creation syntax
- You can take advantage of this behavior to create true copies of your list objects, as in the following example:

```python
mylist1=[1,2,3] # this creates a list object
mylist2=[]+mylist1 # this creates a second, distinct list object (with the same structure as the first)
mylist1[0]=999 # therefore this only changes the first object
print(mylist1)
print(mylist2)
```

## Creating a empty list

Either of the following creates an empty list with no elements in it.

```python
mylist1=[]
mylist2=list()
print(mylist1)
print(mylist2)
```

Sometimes you want to create a list with a certain number of identical elements. You can do this with the "\*" operator. The following creates a list of 10 zeros.

```python
mylist=[0]*10
print(mylist)
```

## Drawbacks to Using "for" Loops with "in" to iterate through a list

- A "for" loop is a convenient way to iterate through a list
- The loop target variable assumes the value of each list item in turn
- However, you can't change the list this way. See the following example:

```python
mylist=[1,2,3]
for n in mylist:
  n=n*5
# no effect on the list!
print(mylist)
```

## Changing List Items

- In order to change a list item, you need to know the index of the item you wish to change. For example:

```python
mylist=[1,2,3]
mylist[1]=444
print(mylist)
```

## Iterating Over a List Using Index Values

There are two techniques for iterating over a list using index values. They are:

- Setting up a counter variable outside the loop and continually updating the variable as you move to the next position in the list
- Using the range() function to create a custom range that represents the size of your list

## Using a Counter Variable

You can add a counter variable to your "for" loop to keep track of the list index. The following code doubles each element in the list.

```python
mylist=[1,2,3]
counter=0
for _ in mylist:
  mylist[counter]*=2
  counter+=1
print(mylist)
```

## Using the range Function

The range function will also generate all of the indeces of the list, when combined with the len function. The following code is another way to double all of the elements of a list.

```python
mylist=[1,2,3]
for i in range(len(mylist)):
  mylist[i]*=2
print(mylist)
```

## Slicing Lists

- You can extract multiple items from a list.
- This is done with slicing; the syntax for this is the same as what we saw for strings.
- As was the case with strings, the slicing syntax is similar to the range function

```python
names=["Amy","Bob","Carol","Dave","Eddie","Flo"]
someNames=names[1:4]
print(someNames)

moreNames=names[-2:]
print(moreNames)

otherNames=names[0:-1:2]
print(otherNames)
```

## The "in" Operator for Lists

The "in" operator returns True or False depending on whether it finds an item in a list.

```python
foods=["pie","cake","pizza"]
if "lasagne" in foods:
  print("'lasagne' found")
else:
  print("'lasagne' not found")
if "pizza" in foods:
  print("'pizza' found")
else:
  print("'pizza' not found")
```

## The "index" Method

The index method gives the index of an item in a list. If the item is not in the list, it gives an error.

```python
foods=["pie","cake","pizza"]
print(foods.index('pizza'))
# this would generate an error
# print(foods.index('lasagne'))
```

## Sorting a list

You can use the sort method to sort a list (numerically or alphabetically). It replaces the list with the sorted version.

```python
fruits=['melon','orange','banana']
fruits.append('apple')
print(fruits)
fruits.sort()
print(fruits)
sequence=list(range(10,0,-1))
print(sequence)
sequence.sort()
print(sequence)
```

## The max and min Functions

You can use the max and min functions to find the largest and smallest elements of a list.

```python
prices = [3.99, 2.99, 1.99]
print(f"The highest price is ${max(prices)} and the lowest price is ${min(prices)}.")
```

## Removing an Item

You can use the "remove" method to remove an item from the list.

It will generate an error if it is not in the list, so you might want to check if it is using the "in" operator first.

```python
prices=[3.99,2.99,1.99]
prices.remove(2.99)
print(prices)
```

## Removing an Item by its Position

You can also use the "del" keyword to remove an item by specifying its index position.

```python
prices=[3.99,2.99,1.99]
del prices[1]
print(prices)
```

## Reversing a List

You can use the "reverse" method on list to reverse the order of its elements.

This will not sort the list, but will just reverse the order.

```python
names=["Alan","Betty","Dave","Chris"]
names.reverse()
print(names)
```
