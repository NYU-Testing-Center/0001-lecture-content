## The Software Development Process

#### Defining Requirements

The software development process starts with gathering requirements. This involves meeting with the ultimate end users of the software and finding out what they need, then documenting these needs in English.

For instance, we might be developing a database application for theater schools for kids. These schools might need to sell tickets to shows that the schools stage with the kids, track students, track their parents, track donors (for non-profit schools), track the classes that the kids are enrolled in, track expenses and payroll, etc. All of these things are specified in the requirements document, written in English.

After the requirements are developed, they are passed to the programmer to implement in code.

#### Testing

After the first version of the program is completed, it is passed to "alpha" (internal) testers.

As bugs are found, they are passed back to the developer for fixes. There are two main types of bugs: failure to meet the requirements, and meeting the requirements incorrectly.

After alpha testing, the program proceeds to public beta testing. More bugs are collected and fixed. Some programs remain in beta a long time. For instance, Gmail was in beta from 2004 to 2009.

After beta testing, the first official non-testing version-- the 1.0 version-- is released. Feature additions are associated with new version numbers. Major revisions are associated with full increments in the version, from 1.0 to 2.0, say. Most actively-used software is a constant work-in-progress.

## Python: Functions

We begin our study of Python with the study of Python functions.

- A function is a piece of computer code that performs a particular action or set of actions
- Python comes with a large number of built-in functions and you can write your own (we will be learning how later in the semester)
- Functions always begin with a keyword followed by a pair of parenthesis, for example, print()
- You send one or more "arguments" into the function by putting them in the parentheses.
  - For instance, print("hello world")
- Different functions expect different types and number of arguments
- When you ask Python to run a function, you say you have "called" the function

Here is a bit of Python code that calls the print function three times.

````python
print("goodbye","cruel","world", sep='! ')
print()
print('hi I\'m back')
````

## Variables in Python

You can think of a variable as a box that holds a value. The following are "assignment statements" that assign values to the variables city and population using the "=" symbol. One has a string value and one has an integer value.

````python
city="New York City"
population=8419000
print("The population of", city, "is", population)
````

You can print out multiple variables and fixed strings in a single print statement, as below.

You can also change the value of a variable. That is why it is called a variable, since its value can change (vary) as a program runs.

````python
city="New York City"
population=8419000
print("The population of",city,"is",population)
city="Boston"
population=684000
print("The population of", city,"is",population)
````

## Swapping the Values of Two Variables

What if we want to swap the values of two variables? We need a third variable as a temporary holder of one of the values.

After we use the temporary third variable, we can use the "del" keyword to remove it.

````python
v1=10
v2=20
print(v1,v2)
tempvar=v1
v1=v2
v2=tempvar
# remove the temporary variable
del tempvar
print(v1,v2)
````

Note that if you try to print a variable that has been deleted, you will get an error, as in the following:

````python
myvar=10
print(myvar)
del myvar
print(myvar)
````

## Initializing Multiple Variables to the Same Value

We can initialize several variables to the same value if we want in a single line of code, as follows:

````python
i=j=k=14
print(i,j,k)
````

## Variable Names in Python

Variable names in Python must contain only letters (upper and lower case), digits, or underscores. They cannot begin with a digit, but have to begin with either a letter or an underscore. Variable names can't contain spaces. The following are some valid variable names:

````python
population_1961=2000000
_myvar=2.1
x99="hello"
print(population_1961,_myvar,x99)
````

## Python's Case Sensitivity

Python is case-sensitive. So the variables "population\_1961" and "Population\_1961" are distinct variables (as would be POPULATION\_1961).

````python
population_1961=2000000
Population_1961=3000000
print(population_1961,Population_1961)
````

## What is a Variable, Really?

Although you can think of a variable as a box containing a value, more accurately a variable is actually a pointer to an object. So in the statement:

```text
population_1961=2000000
```

Python creates an object which is the integer 2000000 in its memory, and then creates a variable population\_1961 which points to that object.

## Mathematical Expressions

You can do mathematical calculations in Python. The following math operators exist:

- addition: +
- subtraction: -
- multiplication: \*
- division (floating point): /
- division (integer): //
- exponentiation: \*\*
- modulus: %

The following are some mathematical expressions using these operators (and parentheses)

````python
sum=4+4.5 # 8.5
print(sum)
square=3**2 # 9
print(square)
print(3+3*5) # 18
print((3+3)*5) # 30
print(7/2) # 3.5
print(7//2) # 3
print(14%5) # 4
total= sum+square # 17.5
print(total)
````

## The Format Function

The format function formats strings and numbers according to a "format string." Let's look at some examples, first with strings, then with integers, and then with floating point numbers:

````python
print("123456789-123456789-123456789-123456789-")
# two strings + a third string
# left justified in a field of width 20, then right justified in a field of width 10
# < means left-justified, and > right-justified
print(format("Hello","<20s"),format("there",">10s")," world",sep="")
````

````python
# integers
print("123456789-123456789-123456789-123456789-")
# left justified, then right justified
print(format(23,"<20d"),format(192,">10d"), sep='')
````

````python
# floating point numbers
print("123456789-123456789-123456789-123456789-")
# left justified, then right justified
x=23.885
print(format(x,"<20.2f"),format(192.1234,">10.3f"), sep='')
print(x)
````
