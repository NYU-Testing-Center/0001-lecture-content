## Decision Structures (If--Elif-Else) and Boolean Logic

So far we have been writing programs that execute one step after another. But it is often necessary to have programs that take different actions based on the data input to them. This is done with a decision structure: an "If" statement.

## An Example: Overtime Pay

For instance, in the U.S., many workers are legally entitled to extra pay if they work more than 40 hours in a week. This is often paid as "time and a half" which means that they get 1.5 times their hourly wage for every hour over 40.

For instance, if your hourly wage is $20, your overtime wage is $30. If you work 60 hours in a week, you earn 40 \* $20 == $800 plus 20 \* $30 == $600, for a total of $1400.

```python
hours=float(input("How many hours did you work last week?"))
wage=float(input("What was your hourly wage?"))
if (hours>40.0):
  # calculate non-overtime and overtime pay
  non_overtime=40*wage
  overtime=(hours-40)*(wage*1.5)
if (hours<=40.0):
  # otherwise no overtime
  non_overtime=hours*wage
  overtime=0
total_pay=non_overtime+overtime
print(f"Your pay should be ${total_pay:.2f} of which ${overtime:.2f} was overtime pay.")
```

## "If" statements

"If" statements are an example of a selection statement. In the overtime example, the program selects and executes the first group of statements after the "if" when the expression following the first "if" is true and the second group when the expression following the second "if" is true. Since only one can be true, only one of these groups of statements are executed.

The expression after the "if" is called a Boolean expression because it takes on a value of True or False. Here the expression following the first "if" is "(hours>40)". This will be True if the variable "hours" (input by the user) is greater than 40 and False otherwise.

Boolean expressions are named after the nineteenth-century English mathematician George Boole who developed [Boolean algebra](https://www.google.com/url?q=https%3A%2F%2Fen.wikipedia.org%2Fwiki%2FBoolean_algebra), which defined the True and False values and the AND, OR and NOT operations that underpin modern computing and the logic used in all computer languages, including Python. In the computer, True is represented by a value of 1 and False with a value of 0.

## Boolean Expressions

- Used as the condition in the "If" statement and other statements that do logical tests that we will see later on (e.g. "while")
- Usually formed using "relational operators" like less than ("<") or greater than which allow you to compare two values and get a result.

```python
hours=float(input("How many hours did you work last week?"))
# (hours>40) is a Boolean expression
# overtimeFlag is a Boolean variable which takes on the value of True or False
overtimeFlag=(hours>40)
print("Did you work overtime?",overtimeFlag)

if overtimeFlag:
  print("You worked overtime.")
else:
  print("You didn't work overtime.")
```

## Relational Operators

Produce a value of True or False

- a>b (a is greater than b)
- a<b (a is less than b)
- a==b (a is equal to b) (different from the assignment operator "=")
- a!=b (a is not equal to b)
- a<=b (a is less than or equal to b)
- a>=b (a is greater than or equal to b)

```python
ageSimon=int(input("Enter Simon's age:"))
ageSasha=int(input("Enter Sasha's age:"))
if (ageSimon>ageSasha):
  print("Simon is older than Sasha")
else:
  print("Either Sasha is older than Simon, or they are the same age.")
```

## String Comparison

- So far, we have seen numerical comparisons
- We can also compare strings
- Strings that contain alphabetic characters are compared alphabetically, character by character.
- All uppercase characters are "before" all lowercase characters
  - this is because the [ASCII codes](https://theasciicode.com.ar/) for the uppercase characters are lower than those for the lowercase characters
- Other characters are also compared using their ASCII codes

All four print statements below output TRUE.

```python
print("aardvark"<"zebra")

print("ZEBRA"<"aardvark")

print("Zebra"<"aardvark")

print("ZEbra"<"ZEidenberg")
```

```python
print('A',ord('A'))
print('Z',ord('Z'))
print('a',ord('a'))
print('z',ord('z'))
```

## The upper() and lower() string methods

The upper() and lower() string methods can be used on a string to convert it to upper or lower case. The syntax is to put a dot after the string and then the method name. (A method is a function that is connected to a type of object, here a string.)

```python
a="aardvark"
z="ZEBRA"
print(a, a.upper())
print(z, z.lower())

if (a.upper()<z):
  print(f'{a.upper()} is less than {z}')

# remember all of the lower case characters are greater than all of the upper case characters
if (a>z):
  print(f'{a} is greater than {z}')

if (a<z.lower()):
  print(f'{a} is less than {z.lower()}')
```

## Nested Decision Structures

The following example shows an "if" within an "if", using indentation. The second "if" is only executed if we reach the "else" of the first "if."

```python
# tropical fish thrive between 72 and 86 degrees, inclusive
temp=float(input("Enter the temperature in Fahrenheit:"))
if (temp>86):
  print("Your fish are too hot!")
else:
  if (temp<72):
    print("Your fish are too cold!")
  else:
    print("Your fish are fine.")
```

## If-Elif-Else

Elif is a way to express additional conditions on the same level as the first "if". Elif is short for "else if." An elif is only run if all the previous conditions have failed to be True. The final "else" is only run if all of the previous conditions have failed. You can have as many elifs as you like.

Elifs aren't necessary, because everything could be written with nested ifs as shown previously, but they can simplify code by eliminating the need for nesting.

```python
weight=float(input("Enter the weight of your dog in pounds:"))
if weight<=5:
  dogType="teacup"
elif weight<=12:
  dogType="toy"
elif weight<=24:
  dogType="small"
elif weight<=57:
  dogType="medium"
else:
  dogType="large"
print(f"You have a {dogType} dog.")
if (weight>=1000):
  print("Is it Clifford?")
```

## Logical Operators

These work on Boolean values or expressions

- and: works on two Booleans of which both must be True
- or: works on two Booleans at least one of which must be True
- not: works on one Boolean and flips its value; True becomes False and vice versa

```python
print(True and True) # True
print(True and False) # False
print(True or True) # True
print(True or False) # True
print(False or False) # False
print(not False) # True
print(not True) # False
```

## Fish Temperature: Another Version

Following is another version of the program which asks you what the temperature of your fish tank is. The program tells you if the temperature is in the proper range for tropical fish. It uses a boolean variable fishOK which is true if the temperature is in the proper range.

```python
# tropical fish thrive between 72 and 86 degrees, inclusive
# version using "and"
temp=float(input("Enter the temperature in Fahrenheit:"))
fishOK=(temp>=72) and (temp<=86)
if (fishOK):
  print("Your fish are fine.")
else:
  print("Your fish are too hot or too cold.")
```

## Generating Random Numbers and then Guessing them

To generate random integers, you first import the random module and then you use the randint method. The randint method takes the two limits of the range of random integers you want (inclusively.)

This program picks a random number and then sees if the user can guess it.

```python
import random
# generate a random integer between 1 and 3
secret=random.randint(1,3)
guess=int(input("What is your guess between 1 and 3?"))
if (guess==secret):
  print("You guessed it!")
else:
  print("Sorry, you didn't guess it.")
```
