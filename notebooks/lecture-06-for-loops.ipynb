## Count-Controlled ("For") Loops

So far, we have seen condition-controlled or "while" loops that repeat the code in their block while a Boolean condition remains true.

We also have count-controlled or "for" loops that repeat the code in their block a predetermined and fixed number of times.

The following while loop will print the numbers from 1 to 10 on a single line.

```python
i=1
while (i<=10):
  print(i,"",end="")
  i+=1
print()
```

The following is the equivalent for loop. Note that anything you can do with a for loop you can do with a while loop, but the syntax of a for loop is more concise when you are iterating a fixed number of times as in this case.

It makes use of the range function range(start,stop,step) which generates a range of numbers starting at and including start and ending at but not including stop, in intervals of step. If step is omitted, it is one.

So range(1,11) gives the numbers from 1 to 10 (since 1 is included but 11 is omitted). It is the same as saying range(1,11,1).

```python
# note that the first element of the range is included but the second isn't.
for i in range(1,11):
  print(i,"",end="")
print()
```

## The range() function

- The range() function takes at least one argument. In its simplest form, it takes a single integer.
- The range() function returns an "iterable" which is a sequence that can be looped over. An iterable is not a list but is similar to one.
- When passed a single integer n, the range() function will generate an iterable that will cause the loop variable of a for loop to go from 0 to n-1 in steps of 1. So for instance, range(5) generates the numbers 0, 1, 2, 3, and 4 in sequence.
- you can also call range() with two or three arguments. For instance, range(5,10) gives the numbers 5, 6, 7, 8 and 9 and range(2,8,2) gives the three numbers 2, 4, and 6. The first argument is where you start and the third argument is the "step."

```python
# Range examples
# 0 to 4
print(list(range(5)))
# 5 to 9
print(list(range(5,10)))
```

```python
# More range examples
# 0 to 9
print(list(range(10)))
# 1 to 4
print(list(range(1,5)))
```

```python
# More range examples
# 0 to 8 by 2
print(list(range(0,10,2)))
# 1 to 9 by 2
print(list(range(1,10,2)))
```

```python
# using range for reverse counting!
# from 10 to 2 by -2
print(list(range(10,0,-2)))
```

## Behavior of the "for" loop

- The "for" loop will iterate once for each item in the list it is passed, or once for each value generated by the range function.
- Lists in Python are defined by square brackets; items are separated by a comma. They may contain the same type of objects or a mix of different types of objects.
- The first time through the loop, the loop variable, which is the variable following the keyword "for", will take on the first item in the list or the first value generated by the range function.
- The second time it will take the second value.
- This continues until you finish the list or the end of the numbers generated by the range function. (Note range does not generate a list but it generates something that behaves similarly.)

Lists contain ordered collections of data which can be of the same type or different types. We will be talking about lists more later.

The following loops through the three states in the tri-state area around New York City and prints them out. It uses a list of the states.

```python
# create a list of three strings called "tristate"
triState=["New York","New Jersey","Connecticut"]
for state in triState:
  print(state)
```

## Prime Number Detector

The variable after "for" in the loop is called the target variable or the loop variable. This is often used inside the loop. For instance, the following code finds all of the prime numbers below 50 and prints them.

```python
# print 2, it is prime
print(2,"",end="")
iterations=0
# check odd numbers from 3 to 49
for i in range(3,50,2):
  # assume prime to start
  prime=True
  # divide up to the square root
  # since factors come in pairs where one is less than or equal to the square
  # and one is greater than or equal to it
  for divisor in range(2,int(i**0.5)+1):
    # if it divides evenly it isn't prime
    # we can break out of the inner loop
    iterations+=1
    if (i%divisor==0):
      prime=False
      break
  if prime: print(i,"",end="")
print()
print(iterations)
```

## Notes on Nested Loops

- An inner loop will do all of iterations for every single iteration of an outer loop.
- Each inner loop completes before the outer loop.
- The total number of iterations is the product of the number of iterations of the outer and inner loops.

## A Multiplication Table

Here's an example that uses a nested for loop to print out the multiplication table up to 20. Note that the total number of steps in the nested loop is 20\*20 or 400.

```python
# print header
print(" "*4,end="")
for j in range(1,21):
  print(f"{j:4d}",end="")
print()

for i in range(1,21):
  print(f"{i:4d}",end="")
  for j in range(1,21):
    print(f"{i*j:4d}",end="")
  print()
```

## Summing Up the Digits of a Number

The following code uses a "for" loop to sum up the digits of a number.

```python
# sum up the digits of a number
n=int(input("Enter a number:"))
sum=0
k=0
# convert the number to a string
# and loop through its digits
for c in str(n):
  k+=1
  print(f"Digit #{k} is {c}.")
  # to calculate the sum, convert the character back to an integer
  sum+=int(c)
print(f"The sum of the digits is {sum}.")
```

## Factors of a Number

The following code finds all the factors of all of the numbers up to 30. It uses a nested "for" loop.

```python
for i in range(1,31):
  print(f"Factors of {i}: ",end="")
  # inner loop finds factors
  for j in range(1,i+1):
    # following tests if it goes in evenly
    if (i%j==0): print(j,"",end="")
  print()
```
