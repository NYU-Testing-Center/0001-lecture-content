## Problem: Guess a Secret Number

Python selects a random number between 1 and 5. We have to run the program multiple times until we guess it.

```python
import random
secret=random.randint(1,5)

print("I've picked a secret number between 1 and 5.")
guess=int(input("What is your guess?"))
if guess==secret:
  print("You got it!")
elif (guess>=1) and (guess<=5):
  print("Sorry, you guessed wrong.")
else:
  print("That number wasn't in range.")
```

What if we want to allow them to continue to guess, within a single run of the program? We use a while loop, which allows us to repeat some code as long as a particular condition remains true.

```python
import random
secret=random.randint(1,5)

print("I've picked a secret number between 1 and 5.")
guess=0 # set it to zero so the following line will be true the first time

# this code repeats until the condition is false
# that is,
while (secret!=guess):
  # beginning of indented block of statements associated with the "while"
  guess=int(input("What is your guess?"))
  if (guess>5) or (guess<1):
    print("Your guess wasn't between 1 and 5.")
    print("Try again.")
  elif secret!=guess:
    print("Sorry, you guessed wrong. Try again.")
  # end of statements associated with the while

print("You got it!")
```

## How Do While Loops Work?

1. They start by evaluating the boolean expression that follows the word "while"
2. If the expression is false, the block of statements associated with the while loop is not executed, and the program continues below them.
3. If the expression is true, the block of statements associated with the while loop is executed, and then step (1) above is repeated.

The two types of loops in Python are while loops and for loops. In both types of loops, one pass through the block of statements indented under the loop is called an iteration. For instance, if the block is executed five times, the loop is said to have executed five iterations.

## Infinite loop

In the following example, of an effectively infinite loop, we use an accumulator variable. An accumulator variable i is a variable that is increased inside the loop.

```python
# not an infinite loop, but effectively one.
# initialize i to zero
i=0
# set upper limit to one trillion (10 to the 12th power)
limit=1000000000000
# print it using an f string with comma separators
print(f"{limit:,d}")
while(i<limit):
  # increment i by one
  i+=1
  # print every ten million
  if (i%10000000)==0: print(f"{i:,d}")
```

## Sentinel Values

Sentinel values are special input values that are used as a "sentinel" or signal to indicate that the user wants to exit a loop.

For instance, suppose we are simulating a cash register. We collect product prices until the user enters a zero price as the sentinel. Then we output the total before and after tax.

```python
# set the price something other than zero so that the
# while loop runs the first time:
price=1.0
total=0.0
i=1
while (price!=0.0):
  price=float(input(f"Enter the price of item #{i}; enter zero to stop entering prices:"))
  total+=price
  i+=1
print(f"Your pre-tax total is ${total:,.2f}")
print(f"Your total after tax is ${total*1.08875:,.2f}")
```

## The break statement

The break statement allows you to break out of a loop early. The execution of the loop stops and execution continues with the next statement after the loop.

Here's a program that adds up numbers in succession and then stops when the sum is greater than or equal to 1000.

```python
i=0
total=0
# infinite loop?
while True:
  i+=1
  total+=i
  # not an infinite loop because this gets us out
  if total>1000: break
print(f"Ended after adding the numbers from 1 to {i}; total is {total}.")
```

## The continue statement

The continue statement allows you to go to the next iteration of the loop in the middle of the block of statements in the loop.

For instance, suppose we want to print all of the numbers from 1 to 5, but we want to skip 3. Here's how we can use continue to do that:

```python
i=0
# note the last iteration is when i is initially 4,
# since it then adds one to i to make it 5
# and then
while (i<=4):
  i+=1
  # if i is 3, go back to the top of the loop
  if i==3: continue
  print(i)
```

The following shows two ways to print just the even numbers from 2 to 10.

```python
# printing just the even numbers from two to ten
# method 1
i=0
while (i<10):
  i+=2
  print(i)
```

```python
# printing just the even numbers from two to ten
# method 2
i=0
while (i<10):
  i+=1
  # this allows you to skip the odd numbers using continue
  if (i%2)==1: continue
  print(i)
```

## A Rock, Paper, Scissors Tournament

The following simulates a rock/paper/scissors tournament between you and the computer. The game ends when either side gets three wins.

```python
import random
computerWins=0
userWins=0
while (computerWins<3) and (userWins<3):
  computerChoice=random.randint(1,3)
  userChoice=int(input("Enter your choice: rock (1), paper (2), or scissors (3)"))
  print("Computer chose ",end='')
  if computerChoice==1:
    print("rock.")
  elif computerChoice==2:
    print("paper.")
  else:
    print("scissors.")
  difference=computerChoice-userChoice
  if difference==0:
    print("It's a tie.")
  # computer win scenarios
  # either choice is one greater than user choice
  # scissors cuts paper, paper wraps rock
  # or computer chose rock, user scissors
  # which means the difference is -2
  elif (difference==1) or (difference==-2):
    print("The computer won.")
    computerWins+=1
  # otherwise user won
  else:
    print("You won.")
    userWins+=1
print("\nTournament Results:")
print(f"You won {userWins} times.")
print(f"The computer won {computerWins} times.")
```

## Flipping a Coin Until You Get Five Heads in a Row

How many coin flips does it take until you get five heads in a row? Well, it varies!

```python
import random
headsInARow=0
nFlips=0
headsRequired=5
while headsInARow<headsRequired:
  nFlips+=1
  if random.randint(1,2)==1: # consider a one a head
    headsInARow+=1
    print("H",end="")
  else:
    print("T",end="")
    headsInARow=0
  if (nFlips%30==0): print()
print(f"\nIt took {nFlips} flips to get {headsRequired} heads in a row.")
```

## An Updated Secret Number Guesser

Guess a number between 1 and 1000 that the computer has picked at random. The computer gives you a hint by telling you if you are too high or too low and keeps track of your highest low guess and your lowest high guess, so you can hone in on it using that range.

Approximately how many guesses will it take?

```python
import random
secret=random.randint(1,1000)
lowGuess=0
highGuess=1001
GuessNumber=0
# infinite loop, but exits with a break
# when you guess it
while True:
  GuessNumber+=1
  print("Guess number",GuessNumber)
  print("Current maximum low guess:",lowGuess)
  print("Current minimum high guess:",highGuess)
  guess=int(input("What's your guess?"))
  if guess==secret:
    print("You got it!")
    break
  elif guess<secret:
    print("Too low.")
    lowGuess=max(lowGuess,guess)
  else:
    print("Too high.")
    highGuess=min(highGuess,guess)
```

## The Game of Nim

In the game of Nim, there are two players and initially seven sticks. On each player's turn, they may pick up up to three sticks. The player who picks up the final stick loses.

This can be implemented by a nested while loop as follows.

```python
# initially there are seven sticks
sticksLeft=7
while (sticksLeft>0):
  # start with player 1
  player=1
  while True: # outer infinite while loop that we will break out of
    # print out the number of sticks left
    print(f"There are {sticksLeft} sticks left.")
    # inner while loop- one player's turn
    while True: # inner infinite while loop that we will break out of
      sticksTaken=int(input(f"Player {player}, how many sticks do you want to pick up?"))
      # if they selected a valid number of sticks, 1-3 and no more than the number left
      if (sticksTaken<=sticksLeft) and (sticksTaken>=1) and (sticksTaken<=3):
        # take away the sticks they selected
        sticksLeft-=sticksTaken
        # check if the game is over: if so, set a flag that it is.
        gameOver=(sticksLeft==0)
        # if the game is over print out who lost by taking the last stick
        if gameOver:
          print(f"No sticks left. Player {player} lost.")
        # end the turn of this player by breaking out of the inner while loop
        break
      # otherwise get them to enter their choice again
      else:
       print("Invalid entry. Try again.")
    #change player
    if player==1:
      player=2
    else:
      player=1
    #  if the game is over break out of the outer while loop as well
    if gameOver: break
```

## Leibniz's formula

Leibniz's formula for pi is the following:

4−4/3+4/5−4/7+4/9...

Let's implement this with a while loop up to 1,000,001. And see what the difference is with the value of pi provided by the Python math package.

##### Approximating Pi:

```python
import math
limit=10000001
i=1
piApprox=0
sign=1
while (i<=limit):
  piApprox+=(4*sign/i)
  sign=-sign
  i+=2
print(piApprox)
print(f"{piApprox-math.pi:.7f}")
```

## Another way to look at 𝑒

Another way to look at 𝑒 is as the result of compounding interest. If you have one dollar invested at 100% interest for a year, then after a year you have two dollars. However, if this is compounded twice (50% after six months and another 50% after a year), you get 1.5 squared or $2.25. If it is compounded four times, you get 1.25 to the fourth power, or about $2.44. In the limit, when it is compounded continously (an infinite number of times) you get 𝑒 dollars; 𝑒 is about 2.718 (but is an irrational number with an infinite number of non-repeating digits after the decimal point.)

```python
import math
numberCompounds=int(input("Compound how many times?"))
# 100 percent interest
interestRate=1
# divide the rate by the number of compounds
compoundingRate=1+(interestRate/(numberCompounds))
i=0
# initially we have one (dollar)
initialAmount=1
resultingAmount=initialAmount
# compound numberCompounds times
while (i<numberCompounds):
  resultingAmount*=compoundingRate
  i+=1
print(resultingAmount)
print(math.e)
```
